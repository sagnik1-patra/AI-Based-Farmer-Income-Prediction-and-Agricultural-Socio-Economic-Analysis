```python
# =============================================================================
# AI-BASED FARMER INCOME PREDICTION
# MEMORY-EFFICIENT ARTIFICIAL IMMUNE SYSTEM (AIS)
#
# IMPORTANT FIX:
# -----------------------------------------------------------------------------
# The previous code generated ~40,672 one-hot encoded columns and attempted
# to allocate ~11.6 GB RAM.
#
# This version DOES NOT use the old one-hot ColumnTransformer.
#
# Instead:
#   1. Numeric features -> median imputation + StandardScaler
#   2. Categorical features -> most-frequent imputation + OrdinalEncoder
#   3. ID/high-cardinality identifier-like columns are removed
#   4. Data stored as float32
#   5. AIS performs feature selection
#   6. ANN performs final farmer-income prediction
#
# ALL GENERATED FILES HAVE ais_ PREFIX.
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import gc
import json
import yaml
import pickle
import zipfile
import tempfile
import shutil
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import (
    StandardScaler,
    OrdinalEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    explained_variance_score
)

from sklearn.ensemble import ExtraTreesRegressor


# =============================================================================
# 2. SETTINGS
# =============================================================================

warnings.filterwarnings(
    "ignore"
)

np.random.seed(
    42
)

tf.random.set_seed(
    42
)


# =============================================================================
# 3. PROJECT DIRECTORY
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Farmer Income Prediction and Agricultural Socio-Economic Analysis"
)


DATA_SOURCE = os.path.join(
    PROJECT_DIR,
    "farmer-income-data.zip"
)


TRAIN_FILE = "lte_train.csv"

TEST_FILE = "lte_test.csv"

DICTIONARY_FILE = "lte_dictionary.csv"


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. TARGET
# =============================================================================

TARGET_COLUMN = (
    "Target_Variable/Total Income"
)


# =============================================================================
# 5. OUTPUT FILES
# =============================================================================

AIS_H5_PATH = os.path.join(
    PROJECT_DIR,
    "ais_model.h5"
)


AIS_PKL_PATH = os.path.join(
    PROJECT_DIR,
    "ais_preprocessor.pkl"
)


AIS_YAML_PATH = os.path.join(
    PROJECT_DIR,
    "ais_config.yaml"
)


AIS_JSON_PATH = os.path.join(
    PROJECT_DIR,
    "ais_metadata.json"
)


AIS_SELECTED_FEATURES_PATH = os.path.join(
    PROJECT_DIR,
    "ais_selected_features.csv"
)


AIS_ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_accuracy_graph.png"
)


AIS_COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_comparison_graph.png"
)


AIS_HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "ais_heatmap.png"
)


AIS_RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result.csv"
)


AIS_RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result_graph.png"
)


AIS_PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction.csv"
)


AIS_PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction_graph.png"
)


AIS_HISTORY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_training_history_graph.png"
)


AIS_METRICS_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "ais_model_metrics.csv"
)


AIS_RESIDUAL_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_residual_graph.png"
)


AIS_FEATURE_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_feature_selection_graph.png"
)


AIS_CONVERGENCE_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_convergence_graph.png"
)


AIS_ACTUAL_PRED_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_actual_predicted_comparison_graph.png"
)


# =============================================================================
# 6. START
# =============================================================================

print(
    "=" * 80
)

print(
    "AI-BASED FARMER INCOME PREDICTION"
)

print(
    "MEMORY-EFFICIENT AIS FEATURE SELECTION"
)

print(
    "=" * 80
)


print(
    "\nProject directory:"
)

print(
    PROJECT_DIR
)


# =============================================================================
# 7. FIND FILE IN DIRECTORY
# =============================================================================

def find_file_in_directory(
    directory,
    required_name
):

    for root, dirs, files in os.walk(
        directory
    ):

        for file_name in files:

            if (
                file_name.lower()
                ==
                required_name.lower()
            ):

                return os.path.join(
                    root,
                    file_name
                )

    return None


# =============================================================================
# 8. FIND FILE INSIDE ZIP
# =============================================================================

def find_file_in_zip(
    names,
    required_name
):

    for name in names:

        if (
            os.path.basename(
                name
            ).lower()
            ==
            required_name.lower()
        ):

            return name

    return None


# =============================================================================
# 9. LOAD DATASETS
# =============================================================================

def load_datasets():

    print(
        "\nSearching for dataset files..."
    )


    # =========================================================================
    # CASE 1: DATA_SOURCE IS A DIRECTORY
    # =========================================================================

    if os.path.isdir(
        DATA_SOURCE
    ):

        print(
            "Data source detected as directory."
        )


        train_path = find_file_in_directory(
            DATA_SOURCE,
            TRAIN_FILE
        )


        test_path = find_file_in_directory(
            DATA_SOURCE,
            TEST_FILE
        )


        dictionary_path = find_file_in_directory(
            DATA_SOURCE,
            DICTIONARY_FILE
        )


        if (
            train_path is not None
            and
            test_path is not None
            and
            dictionary_path is not None
        ):

            return (

                pd.read_csv(
                    train_path,
                    low_memory=False
                ),

                pd.read_csv(
                    test_path,
                    low_memory=False
                ),

                pd.read_csv(
                    dictionary_path,
                    low_memory=False
                )
            )


    # =========================================================================
    # CASE 2: ACTUAL ZIP FILE
    # =========================================================================

    if os.path.isfile(
        DATA_SOURCE
    ):

        try:

            valid_zip = zipfile.is_zipfile(
                DATA_SOURCE
            )

        except Exception:

            valid_zip = False


        if valid_zip:

            print(
                "Valid ZIP archive detected."
            )


            try:

                with zipfile.ZipFile(
                    DATA_SOURCE,
                    "r"
                ) as zip_ref:

                    names = zip_ref.namelist()


                    train_name = find_file_in_zip(
                        names,
                        TRAIN_FILE
                    )


                    test_name = find_file_in_zip(
                        names,
                        TEST_FILE
                    )


                    dictionary_name = find_file_in_zip(
                        names,
                        DICTIONARY_FILE
                    )


                    if train_name is None:

                        raise FileNotFoundError(
                            TRAIN_FILE
                        )


                    if test_name is None:

                        raise FileNotFoundError(
                            TEST_FILE
                        )


                    if dictionary_name is None:

                        raise FileNotFoundError(
                            DICTIONARY_FILE
                        )


                    with zip_ref.open(
                        train_name
                    ) as f:

                        train_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        test_name
                    ) as f:

                        test_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        dictionary_name
                    ) as f:

                        dictionary_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    return (
                        train_data,
                        test_data,
                        dictionary_data
                    )


            except PermissionError:

                print(
                    "\nDirect ZIP access denied."
                )

                print(
                    "Creating temporary copy..."
                )


                temp_directory = tempfile.mkdtemp()


                temp_zip = os.path.join(
                    temp_directory,
                    "farmer-income-data.zip"
                )


                shutil.copy2(
                    DATA_SOURCE,
                    temp_zip
                )


                with zipfile.ZipFile(
                    temp_zip,
                    "r"
                ) as zip_ref:

                    names = zip_ref.namelist()


                    train_name = find_file_in_zip(
                        names,
                        TRAIN_FILE
                    )


                    test_name = find_file_in_zip(
                        names,
                        TEST_FILE
                    )


                    dictionary_name = find_file_in_zip(
                        names,
                        DICTIONARY_FILE
                    )


                    with zip_ref.open(
                        train_name
                    ) as f:

                        train_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        test_name
                    ) as f:

                        test_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        dictionary_name
                    ) as f:

                        dictionary_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                shutil.rmtree(
                    temp_directory,
                    ignore_errors=True
                )


                return (
                    train_data,
                    test_data,
                    dictionary_data
                )


    # =========================================================================
    # CASE 3: SEARCH PROJECT DIRECTORY
    # =========================================================================

    print(
        "\nSearching project directory "
        "for extracted CSV files..."
    )


    train_path = find_file_in_directory(
        PROJECT_DIR,
        TRAIN_FILE
    )


    test_path = find_file_in_directory(
        PROJECT_DIR,
        TEST_FILE
    )


    dictionary_path = find_file_in_directory(
        PROJECT_DIR,
        DICTIONARY_FILE
    )


    if train_path is None:

        raise FileNotFoundError(
            f"{TRAIN_FILE} was not found."
        )


    if test_path is None:

        raise FileNotFoundError(
            f"{TEST_FILE} was not found."
        )


    if dictionary_path is None:

        raise FileNotFoundError(
            f"{DICTIONARY_FILE} was not found."
        )


    return (

        pd.read_csv(
            train_path,
            low_memory=False
        ),

        pd.read_csv(
            test_path,
            low_memory=False
        ),

        pd.read_csv(
            dictionary_path,
            low_memory=False
        )
    )


# =============================================================================
# 10. LOAD DATA
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "LOADING DATA"
)

print(
    "=" * 80
)


train_df, test_df, dictionary_df = load_datasets()


print(
    "\nTraining shape:",
    train_df.shape
)


print(
    "Test shape:",
    test_df.shape
)


print(
    "Dictionary shape:",
    dictionary_df.shape
)


# =============================================================================
# 11. CLEAN COLUMN NAMES
# =============================================================================

train_df.columns = [

    str(column).strip()

    for column in train_df.columns
]


test_df.columns = [

    str(column).strip()

    for column in test_df.columns
]


dictionary_df.columns = [

    str(column).strip()

    for column in dictionary_df.columns
]


# =============================================================================
# 12. FIND TARGET IF REQUIRED
# =============================================================================

if TARGET_COLUMN not in train_df.columns:

    possible_targets = [

        column

        for column in train_df.columns

        if (
            "income"
            in str(column).lower()
            and
            "total"
            in str(column).lower()
        )
    ]


    if len(
        possible_targets
    ) == 0:

        raise ValueError(
            "Total Income target column could not be found."
        )


    TARGET_COLUMN = possible_targets[
        0
    ]


print(
    "\nTarget column:"
)

print(
    TARGET_COLUMN
)


# =============================================================================
# 13. CLEAN TARGET
# =============================================================================

train_df[
    TARGET_COLUMN
] = pd.to_numeric(

    train_df[
        TARGET_COLUMN
    ],

    errors="coerce"
)


train_df[
    TARGET_COLUMN
] = train_df[
    TARGET_COLUMN
].replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


train_df = train_df.dropna(
    subset=[
        TARGET_COLUMN
    ]
).reset_index(
    drop=True
)


# =============================================================================
# 14. CREATE X AND Y
# =============================================================================

X = train_df.drop(
    columns=[
        TARGET_COLUMN
    ]
).copy()


y = train_df[
    TARGET_COLUMN
].astype(
    np.float32
).copy()


if TARGET_COLUMN in test_df.columns:

    X_test_external = test_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    ).copy()

else:

    X_test_external = test_df.copy()


# =============================================================================
# 15. ALIGN TEST COLUMNS WITH TRAIN COLUMNS
# =============================================================================

for column in X.columns:

    if column not in X_test_external.columns:

        X_test_external[
            column
        ] = np.nan


X_test_external = X_test_external[
    X.columns
]


# =============================================================================
# 16. IDENTIFY HIGH-CARDINALITY / ID-LIKE COLUMNS
#
# FarmerID is an identifier.
#
# Columns such as CITY, VILLAGE and Location created thousands of dummy
# variables in the old OneHotEncoder pipeline.
#
# Ordinal encoding could technically keep them, but very high-cardinality
# identifiers can still be noisy and unnecessarily increase complexity.
#
# We automatically remove:
#   - explicit IDs
#   - almost-unique categorical columns
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "CHECKING HIGH-CARDINALITY FEATURES"
)

print(
    "=" * 80
)


explicit_id_columns = [
    "FarmerID"
]


drop_columns = []


for column in explicit_id_columns:

    if column in X.columns:

        drop_columns.append(
            column
        )


# =============================================================================
# Automatically detect nearly unique categorical identifiers
# =============================================================================

for column in X.columns:

    if (
        X[
            column
        ].dtype
        ==
        "object"
    ):

        unique_count = X[
            column
        ].nunique(
            dropna=True
        )


        unique_ratio = (
            unique_count
            /
            max(
                len(X),
                1
            )
        )


        # ---------------------------------------------------------------------
        # Drop columns with extremely high cardinality.
        # ---------------------------------------------------------------------

        if (
            unique_count > 10000
            or
            unique_ratio > 0.80
        ):

            if column not in drop_columns:

                drop_columns.append(
                    column
                )


print(
    "\nColumns removed from modeling:"
)


if len(
    drop_columns
) == 0:

    print(
        "None"
    )

else:

    for column in drop_columns:

        print(
            " -",
            column
        )


X = X.drop(
    columns=drop_columns,
    errors="ignore"
)


X_test_external = X_test_external.drop(
    columns=drop_columns,
    errors="ignore"
)


# =============================================================================
# 17. DETECT NUMERIC AND CATEGORICAL FEATURES
# =============================================================================

numeric_columns = X.select_dtypes(
    include=[
        np.number
    ]
).columns.tolist()


categorical_columns = [

    column

    for column in X.columns

    if column not in numeric_columns
]


print(
    "\nFeatures after dropping IDs:",
    X.shape[1]
)


print(
    "Numeric features:",
    len(
        numeric_columns
    )
)


print(
    "Categorical features:",
    len(
        categorical_columns
    )
)


# =============================================================================
# 18. CONVERT NUMERIC COLUMNS
# =============================================================================

for column in numeric_columns:

    X[
        column
    ] = pd.to_numeric(
        X[
            column
        ],
        errors="coerce"
    )


    X_test_external[
        column
    ] = pd.to_numeric(
        X_test_external[
            column
        ],
        errors="coerce"
    )


# =============================================================================
# 19. CLEAN CATEGORICAL VALUES
# =============================================================================

for column in categorical_columns:

    X[
        column
    ] = X[
        column
    ].astype(
        "object"
    )


    X_test_external[
        column
    ] = X_test_external[
        column
    ].astype(
        "object"
    )


# =============================================================================
# 20. TRAIN / VALIDATION SPLIT
# =============================================================================

X_train, X_valid, y_train, y_valid = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42
)


print(
    "\nTraining rows:",
    len(
        X_train
    )
)


print(
    "Validation rows:",
    len(
        X_valid
    )
)


# =============================================================================
# 21. MEMORY-EFFICIENT PREPROCESSING
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "MEMORY-EFFICIENT PREPROCESSING"
)

print(
    "=" * 80
)


# =============================================================================
# 22. NUMERIC IMPUTER
# =============================================================================

numeric_imputer = SimpleImputer(
    strategy="median"
)


if len(
    numeric_columns
) > 0:

    X_train_numeric = numeric_imputer.fit_transform(
        X_train[
            numeric_columns
        ]
    )


    X_valid_numeric = numeric_imputer.transform(
        X_valid[
            numeric_columns
        ]
    )


    X_test_numeric = numeric_imputer.transform(
        X_test_external[
            numeric_columns
        ]
    )


else:

    X_train_numeric = np.empty(
        (
            len(X_train),
            0
        ),
        dtype=np.float32
    )


    X_valid_numeric = np.empty(
        (
            len(X_valid),
            0
        ),
        dtype=np.float32
    )


    X_test_numeric = np.empty(
        (
            len(X_test_external),
            0
        ),
        dtype=np.float32
    )


# =============================================================================
# 23. NUMERIC SCALER
# =============================================================================

numeric_scaler = StandardScaler()


if X_train_numeric.shape[
    1
] > 0:

    X_train_numeric = numeric_scaler.fit_transform(
        X_train_numeric
    )


    X_valid_numeric = numeric_scaler.transform(
        X_valid_numeric
    )


    X_test_numeric = numeric_scaler.transform(
        X_test_numeric
    )


# =============================================================================
# 24. CATEGORICAL IMPUTATION
# =============================================================================

categorical_imputer = SimpleImputer(
    strategy="most_frequent"
)


if len(
    categorical_columns
) > 0:

    X_train_categorical = categorical_imputer.fit_transform(
        X_train[
            categorical_columns
        ]
    )


    X_valid_categorical = categorical_imputer.transform(
        X_valid[
            categorical_columns
        ]
    )


    X_test_categorical = categorical_imputer.transform(
        X_test_external[
            categorical_columns
        ]
    )


else:

    X_train_categorical = np.empty(
        (
            len(X_train),
            0
        ),
        dtype=np.float32
    )


    X_valid_categorical = np.empty(
        (
            len(X_valid),
            0
        ),
        dtype=np.float32
    )


    X_test_categorical = np.empty(
        (
            len(X_test_external),
            0
        ),
        dtype=np.float32
    )


# =============================================================================
# 25. ORDINAL ENCODER
#
# THIS IS THE MAIN MEMORY FIX.
#
# One categorical column = one output column.
#
# CITY no longer becomes thousands of columns.
# VILLAGE no longer becomes thousands of columns.
# =============================================================================

categorical_encoder = OrdinalEncoder(

    handle_unknown="use_encoded_value",

    unknown_value=-1,

    encoded_missing_value=-1
)


if X_train_categorical.shape[
    1
] > 0:

    X_train_categorical = categorical_encoder.fit_transform(
        X_train_categorical
    )


    X_valid_categorical = categorical_encoder.transform(
        X_valid_categorical
    )


    X_test_categorical = categorical_encoder.transform(
        X_test_categorical
    )


# =============================================================================
# 26. CONVERT EVERYTHING TO FLOAT32
# =============================================================================

X_train_numeric = np.asarray(
    X_train_numeric,
    dtype=np.float32
)


X_valid_numeric = np.asarray(
    X_valid_numeric,
    dtype=np.float32
)


X_test_numeric = np.asarray(
    X_test_numeric,
    dtype=np.float32
)


X_train_categorical = np.asarray(
    X_train_categorical,
    dtype=np.float32
)


X_valid_categorical = np.asarray(
    X_valid_categorical,
    dtype=np.float32
)


X_test_categorical = np.asarray(
    X_test_categorical,
    dtype=np.float32
)


# =============================================================================
# 27. COMBINE NUMERIC + CATEGORICAL
# =============================================================================

X_train_processed = np.concatenate(
    [
        X_train_numeric,
        X_train_categorical
    ],
    axis=1
).astype(
    np.float32,
    copy=False
)


X_valid_processed = np.concatenate(
    [
        X_valid_numeric,
        X_valid_categorical
    ],
    axis=1
).astype(
    np.float32,
    copy=False
)


X_test_processed = np.concatenate(
    [
        X_test_numeric,
        X_test_categorical
    ],
    axis=1
).astype(
    np.float32,
    copy=False
)


# =============================================================================
# 28. FINAL FEATURE NAMES
# =============================================================================

processed_feature_names = (
    numeric_columns
    +
    categorical_columns
)


print(
    "\nProcessed training shape:",
    X_train_processed.shape
)


print(
    "Processed validation shape:",
    X_valid_processed.shape
)


print(
    "Processed test shape:",
    X_test_processed.shape
)


print(
    "\nNumber of processed features:",
    X_train_processed.shape[1]
)


# =============================================================================
# 29. MEMORY INFORMATION
# =============================================================================

train_memory_mb = (
    X_train_processed.nbytes
    /
    1024
    /
    1024
)


valid_memory_mb = (
    X_valid_processed.nbytes
    /
    1024
    /
    1024
)


test_memory_mb = (
    X_test_processed.nbytes
    /
    1024
    /
    1024
)


print(
    f"\nTraining matrix memory: "
    f"{train_memory_mb:.2f} MB"
)


print(
    f"Validation matrix memory: "
    f"{valid_memory_mb:.2f} MB"
)


print(
    f"Test matrix memory: "
    f"{test_memory_mb:.2f} MB"
)


print(
    "\nPrevious failed matrix required approximately 11.6 GB."
)


# =============================================================================
# 30. CLEAN TEMPORARY ARRAYS
# =============================================================================

del X_train_numeric
del X_valid_numeric
del X_test_numeric

del X_train_categorical
del X_valid_categorical
del X_test_categorical


gc.collect()


# =============================================================================
# 31. TARGET SCALER
# =============================================================================

target_scaler = StandardScaler()


y_train_array = np.asarray(
    y_train,
    dtype=np.float32
)


y_valid_array = np.asarray(
    y_valid,
    dtype=np.float32
)


y_train_scaled = target_scaler.fit_transform(
    y_train_array.reshape(
        -1,
        1
    )
).flatten().astype(
    np.float32
)


y_valid_scaled = target_scaler.transform(
    y_valid_array.reshape(
        -1,
        1
    )
).flatten().astype(
    np.float32
)


# =============================================================================
# 32. AIS CONFIGURATION
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "AIS CONFIGURATION"
)

print(
    "=" * 80
)


TOTAL_FEATURES = X_train_processed.shape[
    1
]


AIS_POPULATION_SIZE = 20

AIS_GENERATIONS = 15

AIS_CLONE_FACTOR = 4

AIS_MUTATION_RATE = 0.08

AIS_RANDOM_INJECTION = 4


AIS_MIN_FEATURES = min(
    8,
    TOTAL_FEATURES
)


AIS_MAX_FEATURES = min(
    35,
    TOTAL_FEATURES
)


AIS_CANDIDATE_POOL_SIZE = min(
    60,
    TOTAL_FEATURES
)


print(
    "\nTotal processed features:",
    TOTAL_FEATURES
)


print(
    "AIS candidate pool:",
    AIS_CANDIDATE_POOL_SIZE
)


print(
    "Population:",
    AIS_POPULATION_SIZE
)


print(
    "Generations:",
    AIS_GENERATIONS
)


# =============================================================================
# 33. PRE-SCREEN FEATURES USING EXTRATREES
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "CREATING AIS CANDIDATE FEATURE POOL"
)

print(
    "=" * 80
)


screening_model = ExtraTreesRegressor(

    n_estimators=120,

    random_state=42,

    n_jobs=-1,

    max_depth=12,

    max_features="sqrt"
)


screening_model.fit(
    X_train_processed,
    y_train_array
)


feature_importances = screening_model.feature_importances_


candidate_indices = np.argsort(
    feature_importances
)[::-1][
    :AIS_CANDIDATE_POOL_SIZE
]


candidate_feature_names = [

    processed_feature_names[
        index
    ]

    for index in candidate_indices
]


print(
    "\nTop AIS candidate features:"
)


for index, feature in enumerate(
    candidate_feature_names,
    start=1
):

    print(
        f"{index:02d}. {feature}"
    )


# =============================================================================
# 34. CREATE SMALL CANDIDATE MATRICES
# =============================================================================

X_train_candidate = X_train_processed[
    :,
    candidate_indices
].astype(
    np.float32,
    copy=False
)


X_valid_candidate = X_valid_processed[
    :,
    candidate_indices
].astype(
    np.float32,
    copy=False
)


X_test_candidate = X_test_processed[
    :,
    candidate_indices
].astype(
    np.float32,
    copy=False
)


# =============================================================================
# 35. FREE FULL MATRICES WE DON'T NEED DURING AIS
# =============================================================================

# Keep full arrays until final indices are resolved.
# They are small in this version anyway.

gc.collect()


# =============================================================================
# 36. AIS RANDOM GENERATOR
# =============================================================================

rng = np.random.default_rng(
    42
)


# =============================================================================
# 37. CREATE ANTIBODY
# =============================================================================

def create_antibody():

    mask = np.zeros(
        AIS_CANDIDATE_POOL_SIZE,
        dtype=np.int8
    )


    feature_count = int(
        rng.integers(
            AIS_MIN_FEATURES,
            AIS_MAX_FEATURES + 1
        )
    )


    selected = rng.choice(

        AIS_CANDIDATE_POOL_SIZE,

        size=feature_count,

        replace=False
    )


    mask[
        selected
    ] = 1


    return mask


# =============================================================================
# 38. REPAIR ANTIBODY
# =============================================================================

def repair_antibody(
    mask
):

    mask = mask.copy()


    selected = np.where(
        mask == 1
    )[0]


    # =========================================================================
    # TOO FEW
    # =========================================================================

    if len(
        selected
    ) < AIS_MIN_FEATURES:

        available = np.where(
            mask == 0
        )[0]


        needed = (
            AIS_MIN_FEATURES
            -
            len(
                selected
            )
        )


        if len(
            available
        ) > 0:

            additions = rng.choice(

                available,

                size=min(
                    needed,
                    len(
                        available
                    )
                ),

                replace=False
            )


            mask[
                additions
            ] = 1


    # =========================================================================
    # TOO MANY
    # =========================================================================

    selected = np.where(
        mask == 1
    )[0]


    if len(
        selected
    ) > AIS_MAX_FEATURES:

        number_to_remove = (
            len(
                selected
            )
            -
            AIS_MAX_FEATURES
        )


        removals = rng.choice(

            selected,

            size=number_to_remove,

            replace=False
        )


        mask[
            removals
        ] = 0


    return mask


# =============================================================================
# 39. MUTATE ANTIBODY
# =============================================================================

def mutate_antibody(
    antibody,
    mutation_rate
):

    clone = antibody.copy()


    mutation_mask = (

        rng.random(
            AIS_CANDIDATE_POOL_SIZE
        )

        <

        mutation_rate
    )


    clone[
        mutation_mask
    ] = (
        1
        -
        clone[
            mutation_mask
        ]
    )


    clone = repair_antibody(
        clone
    )


    return clone


# =============================================================================
# 40. AIS FITNESS
#
# ExtraTrees is used as a fast surrogate during immune optimization.
#
# Fitness =
#       validation R²
#       -
#       small complexity penalty
# =============================================================================

fitness_cache = {}


def evaluate_antibody(
    antibody
):

    key = tuple(
        antibody.tolist()
    )


    if key in fitness_cache:

        return fitness_cache[
            key
        ]


    selected = np.where(
        antibody == 1
    )[0]


    if len(
        selected
    ) < AIS_MIN_FEATURES:

        return -999.0


    Xtr = X_train_candidate[
        :,
        selected
    ]


    Xva = X_valid_candidate[
        :,
        selected
    ]


    surrogate = ExtraTreesRegressor(

        n_estimators=60,

        random_state=42,

        n_jobs=-1,

        max_depth=10,

        max_features="sqrt"
    )


    surrogate.fit(
        Xtr,
        y_train_array
    )


    prediction = surrogate.predict(
        Xva
    )


    r2 = r2_score(
        y_valid_array,
        prediction
    )


    feature_penalty = (

        len(
            selected
        )

        /

        AIS_CANDIDATE_POOL_SIZE

        *

        0.02
    )


    fitness = (
        r2
        -
        feature_penalty
    )


    fitness_cache[
        key
    ] = fitness


    return fitness


# =============================================================================
# 41. INITIAL POPULATION
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "INITIALIZING AIS POPULATION"
)

print(
    "=" * 80
)


population = [

    create_antibody()

    for _ in range(
        AIS_POPULATION_SIZE
    )
]


# =============================================================================
# 42. RUN AIS
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "RUNNING ARTIFICIAL IMMUNE SYSTEM"
)

print(
    "=" * 80
)


best_antibody = None

best_fitness = -np.inf

ais_fitness_history = []

ais_feature_count_history = []


for generation in range(
    AIS_GENERATIONS
):

    # =========================================================================
    # EVALUATE POPULATION
    # =========================================================================

    scores = [

        evaluate_antibody(
            antibody
        )

        for antibody in population
    ]


    # =========================================================================
    # SORT
    # =========================================================================

    order = np.argsort(
        scores
    )[::-1]


    population = [

        population[
            index
        ]

        for index in order
    ]


    scores = [

        scores[
            index
        ]

        for index in order
    ]


    # =========================================================================
    # GLOBAL BEST
    # =========================================================================

    if scores[
        0
    ] > best_fitness:

        best_fitness = scores[
            0
        ]


        best_antibody = population[
            0
        ].copy()


    ais_fitness_history.append(
        float(
            best_fitness
        )
    )


    ais_feature_count_history.append(
        int(
            best_antibody.sum()
        )
    )


    print(

        f"Generation "
        f"{generation + 1:02d}/"
        f"{AIS_GENERATIONS} | "

        f"Best Fitness = "
        f"{best_fitness:.6f} | "

        f"Selected Features = "
        f"{int(best_antibody.sum())}"
    )


    # =========================================================================
    # ELITE CELLS
    # =========================================================================

    elite_count = max(
        4,
        AIS_POPULATION_SIZE // 4
    )


    elites = population[
        :elite_count
    ]


    elite_scores = scores[
        :elite_count
    ]


    # =========================================================================
    # CLONAL EXPANSION
    # =========================================================================

    clones = []


    for rank, antibody in enumerate(
        elites
    ):

        adaptive_mutation_rate = (

            AIS_MUTATION_RATE

            *

            (
                1
                +
                rank
                /
                elite_count
            )
        )


        number_of_clones = (

            AIS_CLONE_FACTOR

            *

            (
                elite_count
                -
                rank
            )
        )


        for _ in range(
            number_of_clones
        ):

            clone = mutate_antibody(

                antibody,

                adaptive_mutation_rate
            )


            clones.append(
                clone
            )


    # =========================================================================
    # EVALUATE CLONES
    # =========================================================================

    clone_scores = [

        evaluate_antibody(
            clone
        )

        for clone in clones
    ]


    # =========================================================================
    # COMBINE
    # =========================================================================

    combined_population = (
        elites
        +
        clones
    )


    combined_scores = (
        elite_scores
        +
        clone_scores
    )


    combined_order = np.argsort(
        combined_scores
    )[::-1]


    survivor_count = (

        AIS_POPULATION_SIZE

        -

        AIS_RANDOM_INJECTION
    )


    new_population = [

        combined_population[
            index
        ]

        for index in combined_order[
            :survivor_count
        ]
    ]


    # =========================================================================
    # RANDOM IMMUNE CELLS
    # =========================================================================

    for _ in range(
        AIS_RANDOM_INJECTION
    ):

        new_population.append(
            create_antibody()
        )


    population = new_population


    gc.collect()


# =============================================================================
# 43. AIS COMPLETE
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "AIS OPTIMIZATION COMPLETED"
)

print(
    "=" * 80
)


selected_candidate_positions = np.where(
    best_antibody == 1
)[0]


selected_original_indices = candidate_indices[
    selected_candidate_positions
]


selected_feature_names = [

    processed_feature_names[
        index
    ]

    for index in selected_original_indices
]


selected_importances = feature_importances[
    selected_original_indices
]


print(
    "\nBest AIS fitness:",
    best_fitness
)


print(
    "AIS selected features:",
    len(
        selected_feature_names
    )
)


print(
    "\nSelected feature list:"
)


for index, feature in enumerate(
    selected_feature_names,
    start=1
):

    print(
        f"{index:02d}. {feature}"
    )


# =============================================================================
# 44. SAVE SELECTED FEATURES CSV
# =============================================================================

ais_selected_features_df = pd.DataFrame(
    {
        "Feature_Name":
            selected_feature_names,

        "Original_Feature_Index":
            selected_original_indices,

        "Importance":
            selected_importances
    }
)


ais_selected_features_df = (
    ais_selected_features_df
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


ais_selected_features_df.to_csv(
    AIS_SELECTED_FEATURES_PATH,
    index=False
)


# =============================================================================
# 45. CREATE FINAL AIS MATRICES
# =============================================================================

X_train_ais = X_train_processed[
    :,
    selected_original_indices
].astype(
    np.float32,
    copy=False
)


X_valid_ais = X_valid_processed[
    :,
    selected_original_indices
].astype(
    np.float32,
    copy=False
)


X_test_ais = X_test_processed[
    :,
    selected_original_indices
].astype(
    np.float32,
    copy=False
)


print(
    "\nFinal AIS training shape:",
    X_train_ais.shape
)


print(
    "Final AIS validation shape:",
    X_valid_ais.shape
)


print(
    "Final AIS test shape:",
    X_test_ais.shape
)


# =============================================================================
# 46. RELEASE LARGE ARRAYS
# =============================================================================

del X_train_candidate
del X_valid_candidate
del X_test_candidate


del screening_model


gc.collect()


# =============================================================================
# 47. BUILD FINAL AIS ANN
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "BUILDING AIS-OPTIMIZED NEURAL NETWORK"
)

print(
    "=" * 80
)


input_dimension = X_train_ais.shape[
    1
]


ais_model = Sequential(
    [

        Input(
            shape=(
                input_dimension,
            )
        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.25
        ),

        Dense(
            64,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.20
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            0.10
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )
    ]
)


ais_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=[
        "mae"
    ]
)


ais_model.summary()


# =============================================================================
# 48. CALLBACKS
# =============================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=15,

    restore_best_weights=True,

    verbose=1
)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=6,

    min_lr=1e-6,

    verbose=1
)


# =============================================================================
# 49. TRAIN MODEL
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "TRAINING AIS-OPTIMIZED MODEL"
)

print(
    "=" * 80
)


history = ais_model.fit(

    X_train_ais,

    y_train_scaled,

    validation_data=(
        X_valid_ais,
        y_valid_scaled
    ),

    epochs=150,

    batch_size=128,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)


# =============================================================================
# 50. TRAIN PREDICTIONS
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "GENERATING AIS PREDICTIONS"
)

print(
    "=" * 80
)


train_prediction_scaled = ais_model.predict(
    X_train_ais,
    batch_size=512,
    verbose=0
)


train_predictions = target_scaler.inverse_transform(
    train_prediction_scaled
).flatten()


# =============================================================================
# 51. VALIDATION PREDICTIONS
# =============================================================================

valid_prediction_scaled = ais_model.predict(
    X_valid_ais,
    batch_size=512,
    verbose=0
)


valid_predictions = target_scaler.inverse_transform(
    valid_prediction_scaled
).flatten()


# =============================================================================
# 52. TEST PREDICTIONS
# =============================================================================

test_prediction_scaled = ais_model.predict(
    X_test_ais,
    batch_size=512,
    verbose=0
)


test_predictions = target_scaler.inverse_transform(
    test_prediction_scaled
).flatten()


# =============================================================================
# 53. TRAIN METRICS
# =============================================================================

train_mae = mean_absolute_error(
    y_train_array,
    train_predictions
)


train_mse = mean_squared_error(
    y_train_array,
    train_predictions
)


train_rmse = np.sqrt(
    train_mse
)


train_r2 = r2_score(
    y_train_array,
    train_predictions
)


train_ev = explained_variance_score(
    y_train_array,
    train_predictions
)


# =============================================================================
# 54. VALIDATION METRICS
# =============================================================================

valid_mae = mean_absolute_error(
    y_valid_array,
    valid_predictions
)


valid_mse = mean_squared_error(
    y_valid_array,
    valid_predictions
)


valid_rmse = np.sqrt(
    valid_mse
)


valid_r2 = r2_score(
    y_valid_array,
    valid_predictions
)


valid_ev = explained_variance_score(
    y_valid_array,
    valid_predictions
)


# =============================================================================
# 55. REGRESSION SCORE
# =============================================================================

train_range = (

    np.max(
        y_train_array
    )

    -

    np.min(
        y_train_array
    )
)


valid_range = (

    np.max(
        y_valid_array
    )

    -

    np.min(
        y_valid_array
    )
)


if train_range > 0:

    train_score = (

        1

        -

        train_rmse
        /
        train_range
    )

else:

    train_score = 0.0


if valid_range > 0:

    valid_score = (

        1

        -

        valid_rmse
        /
        valid_range
    )

else:

    valid_score = 0.0


train_score = np.clip(
    train_score,
    0,
    1
)


valid_score = np.clip(
    valid_score,
    0,
    1
)


# =============================================================================
# 56. PRINT PERFORMANCE
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "AIS MODEL PERFORMANCE"
)

print(
    "=" * 80
)


print(
    "\nTRAINING"
)


print(
    f"MAE                 : "
    f"{train_mae:,.4f}"
)


print(
    f"MSE                 : "
    f"{train_mse:,.4f}"
)


print(
    f"RMSE                : "
    f"{train_rmse:,.4f}"
)


print(
    f"R²                  : "
    f"{train_r2:.6f}"
)


print(
    f"Explained Variance  : "
    f"{train_ev:.6f}"
)


print(
    f"Regression Score    : "
    f"{train_score * 100:.2f}%"
)


print(
    "\nVALIDATION"
)


print(
    f"MAE                 : "
    f"{valid_mae:,.4f}"
)


print(
    f"MSE                 : "
    f"{valid_mse:,.4f}"
)


print(
    f"RMSE                : "
    f"{valid_rmse:,.4f}"
)


print(
    f"R²                  : "
    f"{valid_r2:.6f}"
)


print(
    f"Explained Variance  : "
    f"{valid_ev:.6f}"
)


print(
    f"Regression Score    : "
    f"{valid_score * 100:.2f}%"
)


# =============================================================================
# 57. SAVE H5
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "SAVING AIS MODEL"
)

print(
    "=" * 80
)


ais_model.save(
    AIS_H5_PATH
)


print(
    "\nSaved:"
)

print(
    AIS_H5_PATH
)


# =============================================================================
# 58. SAVE PKL
# =============================================================================

ais_preprocessor_package = {

    "numeric_columns":
        numeric_columns,

    "categorical_columns":
        categorical_columns,

    "dropped_columns":
        drop_columns,

    "numeric_imputer":
        numeric_imputer,

    "numeric_scaler":
        numeric_scaler,

    "categorical_imputer":
        categorical_imputer,

    "categorical_encoder":
        categorical_encoder,

    "target_scaler":
        target_scaler,

    "target_column":
        TARGET_COLUMN,

    "processed_feature_names":
        processed_feature_names,

    "selected_feature_names":
        selected_feature_names,

    "selected_feature_indices":
        selected_original_indices,

    "candidate_indices":
        candidate_indices,

    "best_antibody":
        best_antibody,

    "best_fitness":
        float(
            best_fitness
        )
}


with open(
    AIS_PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        ais_preprocessor_package,
        file
    )


print(
    "Saved:"
)

print(
    AIS_PKL_PATH
)


# =============================================================================
# 59. SAVE YAML
# =============================================================================

yaml_data = {

    "project": {

        "name":
            "AI-Based Farmer Income Prediction and "
            "Agricultural Socio-Economic Analysis",

        "method":
            "Artificial Immune System Feature Selection",

        "target":
            str(
                TARGET_COLUMN
            ),

        "problem":
            "Regression"
    },

    "preprocessing": {

        "encoding":
            "OrdinalEncoder",

        "numeric_scaling":
            "StandardScaler",

        "numeric_imputation":
            "Median",

        "categorical_imputation":
            "Most Frequent",

        "dtype":
            "float32",

        "processed_feature_count":
            int(
                TOTAL_FEATURES
            )
    },

    "ais": {

        "population":
            int(
                AIS_POPULATION_SIZE
            ),

        "generations":
            int(
                AIS_GENERATIONS
            ),

        "clone_factor":
            int(
                AIS_CLONE_FACTOR
            ),

        "mutation_rate":
            float(
                AIS_MUTATION_RATE
            ),

        "candidate_pool":
            int(
                AIS_CANDIDATE_POOL_SIZE
            ),

        "selected_features":
            int(
                len(
                    selected_feature_names
                )
            ),

        "best_fitness":
            float(
                best_fitness
            )
    },

    "validation_metrics": {

        "MAE":
            float(
                valid_mae
            ),

        "MSE":
            float(
                valid_mse
            ),

        "RMSE":
            float(
                valid_rmse
            ),

        "R2":
            float(
                valid_r2
            ),

        "Explained_Variance":
            float(
                valid_ev
            )
    }
}


with open(
    AIS_YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False
    )


# =============================================================================
# 60. SAVE JSON
# =============================================================================

json_data = {

    "project":
        "AI-Based Farmer Income Prediction",

    "optimization":
        "Artificial Immune System",

    "memory_efficient":
        True,

    "encoding":
        "Ordinal Encoding",

    "original_features":
        int(
            len(
                X.columns
            )
        ),

    "processed_features":
        int(
            TOTAL_FEATURES
        ),

    "ais_candidate_features":
        int(
            AIS_CANDIDATE_POOL_SIZE
        ),

    "ais_selected_features":
        int(
            len(
                selected_feature_names
            )
        ),

    "selected_feature_names": [

        str(feature)

        for feature in selected_feature_names
    ],

    "best_ais_fitness":
        float(
            best_fitness
        ),

    "training_metrics": {

        "MAE":
            float(
                train_mae
            ),

        "RMSE":
            float(
                train_rmse
            ),

        "R2":
            float(
                train_r2
            )
    },

    "validation_metrics": {

        "MAE":
            float(
                valid_mae
            ),

        "RMSE":
            float(
                valid_rmse
            ),

        "R2":
            float(
                valid_r2
            )
    }
}


with open(
    AIS_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4
    )


# =============================================================================
# 61. MODEL METRICS CSV
# =============================================================================

metrics_df = pd.DataFrame(
    {
        "Dataset": [
            "Training",
            "Validation"
        ],

        "MAE": [
            train_mae,
            valid_mae
        ],

        "MSE": [
            train_mse,
            valid_mse
        ],

        "RMSE": [
            train_rmse,
            valid_rmse
        ],

        "R2": [
            train_r2,
            valid_r2
        ],

        "Explained_Variance": [
            train_ev,
            valid_ev
        ],

        "Regression_Score_Percent": [
            train_score * 100,
            valid_score * 100
        ]
    }
)


metrics_df.to_csv(
    AIS_METRICS_CSV_PATH,
    index=False
)


# =============================================================================
# 62. AIS ACCURACY / PERFORMANCE GRAPH
# =============================================================================

print(
    "\nCreating ais_accuracy_graph.png..."
)


metric_names = [
    "R²",
    "Explained\nVariance",
    "Regression\nScore"
]


training_scores = [

    np.clip(
        train_r2,
        0,
        1
    ) * 100,

    np.clip(
        train_ev,
        0,
        1
    ) * 100,

    train_score * 100
]


validation_scores = [

    np.clip(
        valid_r2,
        0,
        1
    ) * 100,

    np.clip(
        valid_ev,
        0,
        1
    ) * 100,

    valid_score * 100
]


x_axis = np.arange(
    len(
        metric_names
    )
)


width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


bars1 = plt.bar(

    x_axis - width / 2,

    training_scores,

    width,

    label="Training"
)


bars2 = plt.bar(

    x_axis + width / 2,

    validation_scores,

    width,

    label="Validation"
)


plt.xlabel(
    "Performance Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "AIS Farmer Income Prediction Performance"
)


plt.xticks(
    x_axis,
    metric_names
)


plt.ylim(
    0,
    105
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        value = bar.get_height()


        plt.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            value + 1,

            f"{value:.1f}%",

            ha="center",

            fontsize=9
        )


plt.tight_layout()


plt.savefig(
    AIS_ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 63. COMPARISON GRAPH
# =============================================================================

print(
    "Creating ais_comparison_graph.png..."
)


metrics = [
    "MAE",
    "RMSE"
]


training_errors = [
    train_mae,
    train_rmse
]


validation_errors = [
    valid_mae,
    valid_rmse
]


x_axis = np.arange(
    len(
        metrics
    )
)


width = 0.35


plt.figure(
    figsize=(
        9,
        6
    )
)


plt.bar(

    x_axis - width / 2,

    training_errors,

    width,

    label="Training"
)


plt.bar(

    x_axis + width / 2,

    validation_errors,

    width,

    label="Validation"
)


plt.xlabel(
    "Regression Metric"
)


plt.ylabel(
    "Error"
)


plt.title(
    "AIS Training vs Validation Error"
)


plt.xticks(
    x_axis,
    metrics
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    AIS_COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 64. CORRELATION HEATMAP
# =============================================================================

print(
    "Creating ais_heatmap.png..."
)


numeric_heatmap_df = train_df.select_dtypes(
    include=[
        np.number
    ]
).copy()


numeric_heatmap_df = numeric_heatmap_df.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


numeric_heatmap_df = numeric_heatmap_df.dropna(
    axis=1,
    how="all"
)


if TARGET_COLUMN in numeric_heatmap_df.columns:

    target_correlations = (
        numeric_heatmap_df
        .corr()[
            TARGET_COLUMN
        ]
        .abs()
        .sort_values(
            ascending=False
        )
    )


    heatmap_columns = (
        target_correlations
        .head(12)
        .index
        .tolist()
    )


else:

    heatmap_columns = (
        numeric_heatmap_df
        .columns[:12]
        .tolist()
    )


correlation_matrix = numeric_heatmap_df[
    heatmap_columns
].corr()


fig, ax = plt.subplots(
    figsize=(
        14,
        11
    )
)


image = ax.imshow(

    correlation_matrix.values,

    vmin=-1,

    vmax=1,

    aspect="auto"
)


colorbar = fig.colorbar(
    image,
    ax=ax
)


colorbar.set_label(
    "Correlation"
)


ax.set_xticks(
    np.arange(
        len(
            heatmap_columns
        )
    )
)


ax.set_yticks(
    np.arange(
        len(
            heatmap_columns
        )
    )
)


ax.set_xticklabels(
    heatmap_columns,
    rotation=90,
    fontsize=8
)


ax.set_yticklabels(
    heatmap_columns,
    fontsize=8
)


for row in range(
    len(
        heatmap_columns
    )
):

    for column in range(
        len(
            heatmap_columns
        )
    ):

        value = correlation_matrix.iloc[
            row,
            column
        ]


        if not np.isnan(
            value
        ):

            ax.text(

                column,

                row,

                f"{value:.2f}",

                ha="center",

                va="center",

                fontsize=7
            )


ax.set_title(
    "AIS Farmer Income Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(
    AIS_HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 65. RESULT CSV
# =============================================================================

result_df = pd.DataFrame(
    {
        "Actual_Income":
            y_valid_array,

        "AIS_Predicted_Income":
            valid_predictions
    }
)


result_df[
    "Residual"
] = (

    result_df[
        "Actual_Income"
    ]

    -

    result_df[
        "AIS_Predicted_Income"
    ]
)


result_df[
    "Absolute_Error"
] = np.abs(
    result_df[
        "Residual"
    ]
)


actual_absolute = np.abs(
    result_df[
        "Actual_Income"
    ].values
)


result_df[
    "Percentage_Error"
] = np.where(

    actual_absolute > 1e-8,

    (
        result_df[
            "Absolute_Error"
        ].values

        /

        actual_absolute
    )

    *

    100,

    np.nan
)


result_df.to_csv(
    AIS_RESULT_CSV_PATH,
    index=False
)


print(
    "\nFirst 10 AIS validation results:"
)


print(
    result_df.head(
        10
    ).to_string(
        index=False
    )
)


# =============================================================================
# 66. RESULT GRAPH
# =============================================================================

print(
    "Creating ais_result_graph.png..."
)


plt.figure(
    figsize=(
        10,
        7
    )
)


plt.scatter(

    result_df[
        "Actual_Income"
    ],

    result_df[
        "AIS_Predicted_Income"
    ],

    alpha=0.45,

    s=20
)


minimum_value = min(

    result_df[
        "Actual_Income"
    ].min(),

    result_df[
        "AIS_Predicted_Income"
    ].min()
)


maximum_value = max(

    result_df[
        "Actual_Income"
    ].max(),

    result_df[
        "AIS_Predicted_Income"
    ].max()
)


plt.plot(

    [
        minimum_value,
        maximum_value
    ],

    [
        minimum_value,
        maximum_value
    ],

    linestyle="--",

    linewidth=2,

    label="Perfect Prediction"
)


plt.xlabel(
    "Actual Farmer Income"
)


plt.ylabel(
    "AIS Predicted Farmer Income"
)


plt.title(
    "AIS Actual vs Predicted Farmer Income"
)


plt.legend()


plt.grid(
    alpha=0.25
)


performance_text = (

    f"R² = {valid_r2:.4f}\n"

    f"MAE = {valid_mae:,.2f}\n"

    f"RMSE = {valid_rmse:,.2f}"
)


plt.text(

    0.05,

    0.95,

    performance_text,

    transform=plt.gca().transAxes,

    verticalalignment="top",

    bbox=dict(
        boxstyle="round",
        alpha=0.2
    )
)


plt.tight_layout()


plt.savefig(
    AIS_RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 67. PREDICTION CSV
# =============================================================================

prediction_df = test_df.copy()


if TARGET_COLUMN in prediction_df.columns:

    prediction_df = prediction_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    )


prediction_df[
    "AIS_Predicted_Total_Income"
] = test_predictions


prediction_df.insert(

    0,

    "Prediction_ID",

    np.arange(
        1,
        len(
            prediction_df
        )
        + 1
    )
)


prediction_df.to_csv(
    AIS_PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nFirst 10 AIS test predictions:"
)


print(

    prediction_df[
        [
            "Prediction_ID",
            "AIS_Predicted_Total_Income"
        ]
    ]
    .head(
        10
    )
    .to_string(
        index=False
    )
)


# =============================================================================
# 68. PREDICTION GRAPH
# =============================================================================

print(
    "Creating ais_prediction_graph.png..."
)


sorted_predictions = np.sort(
    test_predictions
)


prediction_numbers = np.arange(
    1,
    len(
        sorted_predictions
    )
    + 1
)


mean_prediction = np.mean(
    test_predictions
)


plt.figure(
    figsize=(
        12,
        6
    )
)


plt.plot(

    prediction_numbers,

    sorted_predictions,

    linewidth=2,

    label="AIS Predicted Income"
)


plt.axhline(

    mean_prediction,

    linestyle="--",

    linewidth=2,

    label=(
        f"Mean = "
        f"{mean_prediction:,.2f}"
    )
)


plt.xlabel(
    "Test Records Sorted by Predicted Income"
)


plt.ylabel(
    "Predicted Farmer Income"
)


plt.title(
    "AIS Farmer Income Prediction Distribution"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    AIS_PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 69. TRAINING HISTORY
# =============================================================================

print(
    "Creating ais_training_history_graph.png..."
)


training_loss = history.history[
    "loss"
]


validation_loss = history.history[
    "val_loss"
]


epochs = np.arange(
    1,
    len(
        training_loss
    )
    + 1
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.plot(

    epochs,

    training_loss,

    linewidth=2,

    label="Training Loss"
)


plt.plot(

    epochs,

    validation_loss,

    linewidth=2,

    label="Validation Loss"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "MSE Loss"
)


plt.title(
    "AIS-Optimized ANN Training History"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    AIS_HISTORY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 70. AIS CONVERGENCE GRAPH
# =============================================================================

print(
    "Creating ais_convergence_graph.png..."
)


generations = np.arange(
    1,
    AIS_GENERATIONS + 1
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.plot(

    generations,

    ais_fitness_history,

    marker="o",

    linewidth=2
)


plt.xlabel(
    "AIS Generation"
)


plt.ylabel(
    "Best Fitness"
)


plt.title(
    "Artificial Immune System Convergence"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    AIS_CONVERGENCE_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 71. AIS FEATURE SELECTION GRAPH
# =============================================================================

print(
    "Creating ais_feature_selection_graph.png..."
)


feature_plot_df = (
    ais_selected_features_df
    .sort_values(
        "Importance",
        ascending=True
    )
)


plt.figure(
    figsize=(
        12,
        max(
            7,
            len(
                feature_plot_df
            ) * 0.35
        )
    )
)


plt.barh(

    feature_plot_df[
        "Feature_Name"
    ],

    feature_plot_df[
        "Importance"
    ]
)


plt.xlabel(
    "ExtraTrees Screening Importance"
)


plt.ylabel(
    "AIS Selected Feature"
)


plt.title(
    "AIS Selected Features for Farmer Income Prediction"
)


plt.tight_layout()


plt.savefig(
    AIS_FEATURE_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 72. RESIDUAL GRAPH
# =============================================================================

print(
    "Creating ais_residual_graph.png..."
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.scatter(

    result_df[
        "AIS_Predicted_Income"
    ],

    result_df[
        "Residual"
    ],

    alpha=0.45,

    s=20
)


plt.axhline(
    0,
    linestyle="--",
    linewidth=2
)


plt.xlabel(
    "AIS Predicted Farmer Income"
)


plt.ylabel(
    "Residual (Actual - Predicted)"
)


plt.title(
    "AIS Residual Analysis"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    AIS_RESIDUAL_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 73. ACTUAL VS PREDICTED SEQUENCE GRAPH
# =============================================================================

print(
    "Creating ais_actual_predicted_comparison_graph.png..."
)


comparison_df = result_df.sort_values(
    "Actual_Income"
).reset_index(
    drop=True
)


record_numbers = np.arange(
    len(
        comparison_df
    )
)


plt.figure(
    figsize=(
        14,
        7
    )
)


plt.plot(

    record_numbers,

    comparison_df[
        "Actual_Income"
    ],

    linewidth=2,

    label="Actual Income"
)


plt.plot(

    record_numbers,

    comparison_df[
        "AIS_Predicted_Income"
    ],

    linewidth=2,

    label="AIS Predicted Income"
)


plt.xlabel(
    "Validation Records Sorted by Actual Income"
)


plt.ylabel(
    "Farmer Income"
)


plt.title(
    "AIS Actual vs Predicted Farmer Income Comparison"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    AIS_ACTUAL_PRED_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


# =============================================================================
# 74. TEST PREDICTION SUMMARY
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "AIS TEST PREDICTION SUMMARY"
)

print(
    "=" * 80
)


print(
    f"\nNumber of predictions : "
    f"{len(test_predictions):,}"
)


print(
    f"Minimum prediction    : "
    f"{np.min(test_predictions):,.2f}"
)


print(
    f"Maximum prediction    : "
    f"{np.max(test_predictions):,.2f}"
)


print(
    f"Mean prediction       : "
    f"{np.mean(test_predictions):,.2f}"
)


print(
    f"Median prediction     : "
    f"{np.median(test_predictions):,.2f}"
)


print(
    f"Standard deviation    : "
    f"{np.std(test_predictions):,.2f}"
)


# =============================================================================
# 75. VERIFY GENERATED FILES
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "VERIFYING AIS OUTPUT FILES"
)

print(
    "=" * 80
)


generated_files = [

    AIS_H5_PATH,

    AIS_PKL_PATH,

    AIS_YAML_PATH,

    AIS_JSON_PATH,

    AIS_SELECTED_FEATURES_PATH,

    AIS_ACCURACY_GRAPH_PATH,

    AIS_COMPARISON_GRAPH_PATH,

    AIS_HEATMAP_PATH,

    AIS_RESULT_CSV_PATH,

    AIS_RESULT_GRAPH_PATH,

    AIS_PREDICTION_CSV_PATH,

    AIS_PREDICTION_GRAPH_PATH,

    AIS_HISTORY_GRAPH_PATH,

    AIS_METRICS_CSV_PATH,

    AIS_RESIDUAL_GRAPH_PATH,

    AIS_FEATURE_GRAPH_PATH,

    AIS_CONVERGENCE_GRAPH_PATH,

    AIS_ACTUAL_PRED_GRAPH_PATH
]


for path in generated_files:

    if os.path.isfile(
        path
    ):

        file_size_kb = (

            os.path.getsize(
                path
            )

            /

            1024
        )


        print(

            f"[OK] "

            f"{os.path.basename(path)} "

            f"({file_size_kb:,.2f} KB)"
        )


    else:

        print(

            f"[ERROR] "

            f"{os.path.basename(path)}"
        )


# =============================================================================
# 76. FINAL SUMMARY
# =============================================================================

print(
    "\n" + "=" * 80
)

print(
    "AIS PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 80
)


print(
    f"\nFeatures before AIS       : "
    f"{TOTAL_FEATURES}"
)


print(
    f"AIS candidate features    : "
    f"{AIS_CANDIDATE_POOL_SIZE}"
)


print(
    f"AIS selected features     : "
    f"{len(selected_feature_names)}"
)


print(
    f"Best AIS fitness          : "
    f"{best_fitness:.6f}"
)


print(
    "\nVALIDATION PERFORMANCE"
)


print(
    f"MAE                      : "
    f"{valid_mae:,.4f}"
)


print(
    f"MSE                      : "
    f"{valid_mse:,.4f}"
)


print(
    f"RMSE                     : "
    f"{valid_rmse:,.4f}"
)


print(
    f"R²                       : "
    f"{valid_r2:.6f}"
)


print(
    f"Explained Variance       : "
    f"{valid_ev:.6f}"
)


print(
    f"Regression Score         : "
    f"{valid_score * 100:.2f}%"
)


print(
    "\nFiles saved in:"
)

print(
    PROJECT_DIR
)


print(
    """

GENERATED AIS FILES
-------------------

1.  ais_model.h5
2.  ais_preprocessor.pkl
3.  ais_config.yaml
4.  ais_metadata.json
5.  ais_selected_features.csv

6.  ais_accuracy_graph.png
7.  ais_comparison_graph.png
8.  ais_heatmap.png

9.  ais_result.csv
10. ais_result_graph.png

11. ais_prediction.csv
12. ais_prediction_graph.png

13. ais_training_history_graph.png
14. ais_model_metrics.csv
15. ais_residual_graph.png

16. ais_feature_selection_graph.png
17. ais_convergence_graph.png
18. ais_actual_predicted_comparison_graph.png
"""
)


print(
    "=" * 80
)

print(
    "DONE"
)

print(
    "=" * 80
)
```
