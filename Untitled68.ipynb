```python
# =============================================================================
# AI-BASED FARMER INCOME PREDICTION
# COMPLETE STANDALONE RESULTS + VISUALIZATION CODE
#
# This code can be run in a NEW Jupyter cell/kernel.
#
# It:
# 1. Finds/loads lte_train.csv, lte_test.csv and lte_dictionary.csv
# 2. Loads farmer_income_model.h5
# 3. Loads farmer_income_preprocessor.pkl
# 4. Recreates train/validation split
# 5. Preprocesses the data
# 6. Generates train/validation/test predictions
# 7. Calculates regression metrics
# 8. Shows + saves:
#
#       accuracy_graph.png
#       comparison_graph.png
#       heatmap.png
#       result_graph.png
#       prediction_graph.png
#       training_history_graph.png (if history exists; otherwise skipped)
#
# 9. Saves:
#
#       result.csv
#       prediction.csv
#       model_metrics.csv
#
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import json
import pickle
import zipfile
import tempfile
import shutil
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    explained_variance_score
)


warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =============================================================================
# 2. PROJECT DIRECTORY
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Farmer Income Prediction and Agricultural Socio-Economic Analysis"
)


DATA_SOURCE = os.path.join(
    PROJECT_DIR,
    "farmer-income-data.zip"
)


# =============================================================================
# 3. INPUT FILE NAMES
# =============================================================================

TRAIN_FILE = "lte_train.csv"

TEST_FILE = "lte_test.csv"

DICTIONARY_FILE = "lte_dictionary.csv"


# =============================================================================
# 4. SAVED MODEL FILES
# =============================================================================

H5_PATH = os.path.join(
    PROJECT_DIR,
    "farmer_income_model.h5"
)


PKL_PATH = os.path.join(
    PROJECT_DIR,
    "farmer_income_preprocessor.pkl"
)


# =============================================================================
# 5. OUTPUT FILE PATHS
# =============================================================================

ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)


COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "comparison_graph.png"
)


HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)


RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "result.csv"
)


RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)


PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)


PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)


METRICS_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "model_metrics.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 6. START
# =============================================================================

print("=" * 80)

print(
    "AI-BASED FARMER INCOME PREDICTION"
)

print(
    "RESULTS AND VISUALIZATION"
)

print("=" * 80)


print("\nProject directory:")

print(
    PROJECT_DIR
)


# =============================================================================
# 7. CHECK MODEL FILE
# =============================================================================

print("\n" + "=" * 80)
print("CHECKING SAVED MODEL FILES")
print("=" * 80)


if not os.path.isfile(
    H5_PATH
):

    raise FileNotFoundError(
        "\nH5 model was not found:\n"
        f"{H5_PATH}\n\n"
        "Run the model-generation code first."
    )


if not os.path.isfile(
    PKL_PATH
):

    raise FileNotFoundError(
        "\nPKL preprocessor was not found:\n"
        f"{PKL_PATH}\n\n"
        "Run the model-generation code first."
    )


print(
    "\n[OK] H5 model found"
)

print(
    H5_PATH
)


print(
    "\n[OK] PKL preprocessor found"
)

print(
    PKL_PATH
)


# =============================================================================
# 8. LOAD H5 MODEL
# =============================================================================

print("\n" + "=" * 80)
print("LOADING TRAINED MODEL")
print("=" * 80)


model = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)


print(
    "\nModel loaded successfully."
)


model.summary()


# =============================================================================
# 9. LOAD PKL
# =============================================================================

print("\n" + "=" * 80)
print("LOADING PREPROCESSOR")
print("=" * 80)


with open(
    PKL_PATH,
    "rb"
) as file:

    package = pickle.load(
        file
    )


# =============================================================================
# 10. EXTRACT OBJECTS FROM PKL
# =============================================================================

if "preprocessor" not in package:

    raise KeyError(
        "The PKL file does not contain 'preprocessor'."
    )


if "target_scaler" not in package:

    raise KeyError(
        "The PKL file does not contain 'target_scaler'."
    )


preprocessor = package[
    "preprocessor"
]


target_scaler = package[
    "target_scaler"
]


TARGET_COLUMN = package.get(
    "target_column",
    "Target_Variable/Total Income"
)


feature_columns = package.get(
    "feature_columns",
    None
)


print(
    "\nPreprocessor loaded successfully."
)


print(
    "\nTarget variable:"
)

print(
    TARGET_COLUMN
)


if feature_columns is not None:

    print(
        "\nExpected features:",
        len(feature_columns)
    )


# =============================================================================
# 11. HELPER FUNCTION TO FIND FILE IN DIRECTORY
# =============================================================================

def find_file_in_directory(
    directory,
    required_name
):

    for root, dirs, files in os.walk(
        directory
    ):

        for file_name in files:

            if (
                file_name.lower()
                ==
                required_name.lower()
            ):

                return os.path.join(
                    root,
                    file_name
                )

    return None


# =============================================================================
# 12. HELPER FUNCTION TO FIND FILE INSIDE ZIP
# =============================================================================

def find_file_in_zip(
    names,
    required_name
):

    for name in names:

        if (
            os.path.basename(
                name
            ).lower()
            ==
            required_name.lower()
        ):

            return name

    return None


# =============================================================================
# 13. DATA LOADING FUNCTION
# =============================================================================

def load_datasets():

    print(
        "\nSearching for dataset files..."
    )


    # =========================================================================
    # OPTION 1:
    # farmer-income-data.zip is actually a directory
    # =========================================================================

    if os.path.isdir(
        DATA_SOURCE
    ):

        print(
            "\nData source is a directory."
        )


        train_path = find_file_in_directory(
            DATA_SOURCE,
            TRAIN_FILE
        )


        test_path = find_file_in_directory(
            DATA_SOURCE,
            TEST_FILE
        )


        dictionary_path = find_file_in_directory(
            DATA_SOURCE,
            DICTIONARY_FILE
        )


        if (
            train_path
            and test_path
            and dictionary_path
        ):

            train_data = pd.read_csv(
                train_path,
                low_memory=False
            )


            test_data = pd.read_csv(
                test_path,
                low_memory=False
            )


            dictionary_data = pd.read_csv(
                dictionary_path,
                low_memory=False
            )


            return (
                train_data,
                test_data,
                dictionary_data
            )


    # =========================================================================
    # OPTION 2:
    # Real ZIP archive
    # =========================================================================

    if os.path.isfile(
        DATA_SOURCE
    ):

        print(
            "\nData source is a file."
        )


        try:

            is_zip = zipfile.is_zipfile(
                DATA_SOURCE
            )

        except PermissionError:

            is_zip = False


        if is_zip:

            print(
                "Valid ZIP detected."
            )


            try:

                zip_path_to_use = DATA_SOURCE


                with zipfile.ZipFile(
                    zip_path_to_use,
                    "r"
                ) as zip_ref:

                    names = zip_ref.namelist()


                    train_name = find_file_in_zip(
                        names,
                        TRAIN_FILE
                    )


                    test_name = find_file_in_zip(
                        names,
                        TEST_FILE
                    )


                    dictionary_name = find_file_in_zip(
                        names,
                        DICTIONARY_FILE
                    )


                    if (
                        train_name is None
                        or
                        test_name is None
                        or
                        dictionary_name is None
                    ):

                        raise FileNotFoundError(
                            "Required CSV files are missing "
                            "from the ZIP."
                        )


                    with zip_ref.open(
                        train_name
                    ) as f:

                        train_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        test_name
                    ) as f:

                        test_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        dictionary_name
                    ) as f:

                        dictionary_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    return (
                        train_data,
                        test_data,
                        dictionary_data
                    )


            except PermissionError:

                print(
                    "\nDirect ZIP access blocked."
                )

                print(
                    "Trying temporary copy..."
                )


                temp_directory = tempfile.mkdtemp()


                temp_zip = os.path.join(
                    temp_directory,
                    "farmer-income-data.zip"
                )


                shutil.copy2(
                    DATA_SOURCE,
                    temp_zip
                )


                with zipfile.ZipFile(
                    temp_zip,
                    "r"
                ) as zip_ref:

                    names = zip_ref.namelist()


                    train_name = find_file_in_zip(
                        names,
                        TRAIN_FILE
                    )


                    test_name = find_file_in_zip(
                        names,
                        TEST_FILE
                    )


                    dictionary_name = find_file_in_zip(
                        names,
                        DICTIONARY_FILE
                    )


                    with zip_ref.open(
                        train_name
                    ) as f:

                        train_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        test_name
                    ) as f:

                        test_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        dictionary_name
                    ) as f:

                        dictionary_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    return (
                        train_data,
                        test_data,
                        dictionary_data
                    )


    # =========================================================================
    # OPTION 3:
    # Search entire project folder for extracted CSVs
    # =========================================================================

    print(
        "\nSearching project directory "
        "for extracted CSV files..."
    )


    train_path = find_file_in_directory(
        PROJECT_DIR,
        TRAIN_FILE
    )


    test_path = find_file_in_directory(
        PROJECT_DIR,
        TEST_FILE
    )


    dictionary_path = find_file_in_directory(
        PROJECT_DIR,
        DICTIONARY_FILE
    )


    if train_path is None:

        raise FileNotFoundError(
            f"{TRAIN_FILE} not found."
        )


    if test_path is None:

        raise FileNotFoundError(
            f"{TEST_FILE} not found."
        )


    if dictionary_path is None:

        raise FileNotFoundError(
            f"{DICTIONARY_FILE} not found."
        )


    train_data = pd.read_csv(
        train_path,
        low_memory=False
    )


    test_data = pd.read_csv(
        test_path,
        low_memory=False
    )


    dictionary_data = pd.read_csv(
        dictionary_path,
        low_memory=False
    )


    return (
        train_data,
        test_data,
        dictionary_data
    )


# =============================================================================
# 14. LOAD DATA
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATASETS")
print("=" * 80)


train_df, test_df, dictionary_df = load_datasets()


print(
    "\nTraining shape:",
    train_df.shape
)


print(
    "Test shape:",
    test_df.shape
)


print(
    "Dictionary shape:",
    dictionary_df.shape
)


# =============================================================================
# 15. CLEAN COLUMN NAMES
# =============================================================================

train_df.columns = [
    str(column).strip()
    for column in train_df.columns
]


test_df.columns = [
    str(column).strip()
    for column in test_df.columns
]


dictionary_df.columns = [
    str(column).strip()
    for column in dictionary_df.columns
]


# =============================================================================
# 16. VERIFY TARGET
# =============================================================================

if TARGET_COLUMN not in train_df.columns:

    print(
        "\nSaved target name was not found."
    )

    print(
        "Searching for total income column..."
    )


    possible_targets = [

        column

        for column in train_df.columns

        if (
            "income"
            in
            str(column).lower()
            and
            "total"
            in
            str(column).lower()
        )
    ]


    if len(
        possible_targets
    ) == 0:

        raise ValueError(
            "Total income target could not be found."
        )


    TARGET_COLUMN = possible_targets[0]


print(
    "\nUsing target:"
)

print(
    TARGET_COLUMN
)


# =============================================================================
# 17. CLEAN TARGET
# =============================================================================

train_df[
    TARGET_COLUMN
] = pd.to_numeric(
    train_df[
        TARGET_COLUMN
    ],
    errors="coerce"
)


train_df[
    TARGET_COLUMN
] = train_df[
    TARGET_COLUMN
].replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


train_df = train_df.dropna(
    subset=[
        TARGET_COLUMN
    ]
).reset_index(
    drop=True
)


# =============================================================================
# 18. CREATE FEATURES AND TARGET
# =============================================================================

X = train_df.drop(
    columns=[
        TARGET_COLUMN
    ]
).copy()


y = train_df[
    TARGET_COLUMN
].astype(
    float
).copy()


if TARGET_COLUMN in test_df.columns:

    X_test_external = test_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    ).copy()

else:

    X_test_external = test_df.copy()


# =============================================================================
# 19. REPLACE INFINITY
# =============================================================================

X = X.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


X_test_external = X_test_external.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


# =============================================================================
# 20. USE EXACT FEATURES STORED IN PKL
#
# This is VERY important.
# The loaded preprocessor expects the same features used during training.
# =============================================================================

if feature_columns is not None:

    print(
        "\nAligning features with saved model..."
    )


    # Add missing columns

    for column in feature_columns:

        if column not in X.columns:

            X[
                column
            ] = np.nan


        if column not in X_test_external.columns:

            X_test_external[
                column
            ] = np.nan


    # Select exact features and exact order

    X = X[
        feature_columns
    ]


    X_test_external = X_test_external[
        feature_columns
    ]


print(
    "\nFinal feature count:",
    X.shape[1]
)


# =============================================================================
# 21. RECREATE SAME TRAIN / VALIDATION SPLIT
#
# Original model code:
# test_size = 0.20
# random_state = 42
# =============================================================================

X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print(
    "\nTraining rows:",
    len(
        X_train
    )
)


print(
    "Validation rows:",
    len(
        X_valid
    )
)


# =============================================================================
# 22. PREPROCESS DATA USING LOADED PREPROCESSOR
# =============================================================================

print("\n" + "=" * 80)
print("PREPROCESSING DATA")
print("=" * 80)


X_train_processed = preprocessor.transform(
    X_train
)


X_valid_processed = preprocessor.transform(
    X_valid
)


X_test_processed = preprocessor.transform(
    X_test_external
)


# =============================================================================
# 23. CONVERT TO FLOAT32
# =============================================================================

X_train_processed = np.asarray(
    X_train_processed,
    dtype=np.float32
)


X_valid_processed = np.asarray(
    X_valid_processed,
    dtype=np.float32
)


X_test_processed = np.asarray(
    X_test_processed,
    dtype=np.float32
)


print(
    "\nProcessed training shape:",
    X_train_processed.shape
)


print(
    "Processed validation shape:",
    X_valid_processed.shape
)


print(
    "Processed test shape:",
    X_test_processed.shape
)


# =============================================================================
# 24. CHECK MODEL INPUT SIZE
# =============================================================================

expected_input_size = model.input_shape[-1]


actual_input_size = X_train_processed.shape[1]


print(
    "\nModel expected inputs:",
    expected_input_size
)


print(
    "Actual processed inputs:",
    actual_input_size
)


if (
    expected_input_size
    !=
    actual_input_size
):

    raise ValueError(
        "\nMODEL/PREPROCESSOR MISMATCH\n\n"
        f"Model expects {expected_input_size} features "
        f"but preprocessor produced {actual_input_size}.\n\n"
        "Make sure farmer_income_model.h5 and "
        "farmer_income_preprocessor.pkl were generated "
        "during the same training run."
    )


# =============================================================================
# 25. GENERATE TRAINING PREDICTIONS
# =============================================================================

print("\n" + "=" * 80)
print("GENERATING PREDICTIONS")
print("=" * 80)


print(
    "\nGenerating training predictions..."
)


train_pred_scaled = model.predict(
    X_train_processed,
    verbose=0
)


train_predictions = target_scaler.inverse_transform(
    train_pred_scaled
).flatten()


# =============================================================================
# 26. VALIDATION PREDICTIONS
# =============================================================================

print(
    "Generating validation predictions..."
)


valid_pred_scaled = model.predict(
    X_valid_processed,
    verbose=0
)


valid_predictions = target_scaler.inverse_transform(
    valid_pred_scaled
).flatten()


# =============================================================================
# 27. TEST PREDICTIONS
# =============================================================================

print(
    "Generating test predictions..."
)


test_pred_scaled = model.predict(
    X_test_processed,
    verbose=0
)


test_predictions = target_scaler.inverse_transform(
    test_pred_scaled
).flatten()


# =============================================================================
# 28. ARRAYS
# =============================================================================

y_train_array = np.asarray(
    y_train,
    dtype=float
)


y_valid_array = np.asarray(
    y_valid,
    dtype=float
)


# =============================================================================
# 29. TRAINING METRICS
# =============================================================================

train_mae = mean_absolute_error(
    y_train_array,
    train_predictions
)


train_mse = mean_squared_error(
    y_train_array,
    train_predictions
)


train_rmse = np.sqrt(
    train_mse
)


train_r2 = r2_score(
    y_train_array,
    train_predictions
)


train_ev = explained_variance_score(
    y_train_array,
    train_predictions
)


# =============================================================================
# 30. VALIDATION METRICS
# =============================================================================

valid_mae = mean_absolute_error(
    y_valid_array,
    valid_predictions
)


valid_mse = mean_squared_error(
    y_valid_array,
    valid_predictions
)


valid_rmse = np.sqrt(
    valid_mse
)


valid_r2 = r2_score(
    y_valid_array,
    valid_predictions
)


valid_ev = explained_variance_score(
    y_valid_array,
    valid_predictions
)


# =============================================================================
# 31. REGRESSION PERFORMANCE SCORE
#
# This is NOT classification accuracy.
#
# Score = 1 - Normalized RMSE
# =============================================================================

train_range = (
    np.max(
        y_train_array
    )
    -
    np.min(
        y_train_array
    )
)


valid_range = (
    np.max(
        y_valid_array
    )
    -
    np.min(
        y_valid_array
    )
)


if train_range > 0:

    train_score = (
        1
        -
        train_rmse
        /
        train_range
    )

else:

    train_score = 0


if valid_range > 0:

    valid_score = (
        1
        -
        valid_rmse
        /
        valid_range
    )

else:

    valid_score = 0


train_score = np.clip(
    train_score,
    0,
    1
)


valid_score = np.clip(
    valid_score,
    0,
    1
)


# =============================================================================
# 32. PRINT PERFORMANCE
# =============================================================================

print("\n" + "=" * 80)
print("MODEL PERFORMANCE")
print("=" * 80)


print("\nTRAINING RESULTS")


print(
    f"MAE                 : "
    f"{train_mae:,.4f}"
)


print(
    f"MSE                 : "
    f"{train_mse:,.4f}"
)


print(
    f"RMSE                : "
    f"{train_rmse:,.4f}"
)


print(
    f"R²                  : "
    f"{train_r2:.6f}"
)


print(
    f"Explained Variance  : "
    f"{train_ev:.6f}"
)


print(
    f"Regression Score    : "
    f"{train_score * 100:.2f}%"
)


print("\nVALIDATION RESULTS")


print(
    f"MAE                 : "
    f"{valid_mae:,.4f}"
)


print(
    f"MSE                 : "
    f"{valid_mse:,.4f}"
)


print(
    f"RMSE                : "
    f"{valid_rmse:,.4f}"
)


print(
    f"R²                  : "
    f"{valid_r2:.6f}"
)


print(
    f"Explained Variance  : "
    f"{valid_ev:.6f}"
)


print(
    f"Regression Score    : "
    f"{valid_score * 100:.2f}%"
)


# =============================================================================
# 33. SAVE MODEL METRICS CSV
# =============================================================================

metrics_df = pd.DataFrame(
    {
        "Dataset": [
            "Training",
            "Validation"
        ],

        "MAE": [
            train_mae,
            valid_mae
        ],

        "MSE": [
            train_mse,
            valid_mse
        ],

        "RMSE": [
            train_rmse,
            valid_rmse
        ],

        "R2": [
            train_r2,
            valid_r2
        ],

        "Explained_Variance": [
            train_ev,
            valid_ev
        ],

        "Regression_Score_Percent": [
            train_score * 100,
            valid_score * 100
        ]
    }
)


metrics_df.to_csv(
    METRICS_CSV_PATH,
    index=False
)


print(
    "\nSaved:",
    METRICS_CSV_PATH
)


# =============================================================================
# 34. ACCURACY / PERFORMANCE GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING ACCURACY GRAPH")
print("=" * 80)


metric_names = [
    "R²",
    "Explained\nVariance",
    "Regression\nScore"
]


training_scores = [

    np.clip(
        train_r2,
        0,
        1
    ) * 100,

    np.clip(
        train_ev,
        0,
        1
    ) * 100,

    train_score * 100
]


validation_scores = [

    np.clip(
        valid_r2,
        0,
        1
    ) * 100,

    np.clip(
        valid_ev,
        0,
        1
    ) * 100,

    valid_score * 100
]


x = np.arange(
    len(
        metric_names
    )
)


width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


bars1 = plt.bar(
    x - width / 2,
    training_scores,
    width,
    label="Training"
)


bars2 = plt.bar(
    x + width / 2,
    validation_scores,
    width,
    label="Validation"
)


plt.xlabel(
    "Performance Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "Farmer Income Prediction - Regression Performance"
)


plt.xticks(
    x,
    metric_names
)


plt.ylim(
    0,
    105
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        value = bar.get_height()

        plt.text(
            bar.get_x()
            +
            bar.get_width() / 2,

            value + 1,

            f"{value:.1f}%",

            ha="center",

            fontsize=9
        )


plt.tight_layout()


plt.savefig(
    ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    ACCURACY_GRAPH_PATH
)


# =============================================================================
# 35. COMPARISON GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING COMPARISON GRAPH")
print("=" * 80)


metric_names = [
    "MAE",
    "RMSE"
]


train_values = [
    train_mae,
    train_rmse
]


valid_values = [
    valid_mae,
    valid_rmse
]


x = np.arange(
    len(
        metric_names
    )
)


width = 0.35


plt.figure(
    figsize=(
        9,
        6
    )
)


bars1 = plt.bar(
    x - width / 2,
    train_values,
    width,
    label="Training"
)


bars2 = plt.bar(
    x + width / 2,
    valid_values,
    width,
    label="Validation"
)


plt.xlabel(
    "Error Metric"
)


plt.ylabel(
    "Error Value"
)


plt.title(
    "Training vs Validation Error Comparison"
)


plt.xticks(
    x,
    metric_names
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.25
)


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        value = bar.get_height()

        plt.text(
            bar.get_x()
            +
            bar.get_width() / 2,

            value,

            f"{value:,.0f}",

            ha="center",

            va="bottom",

            fontsize=9
        )


plt.tight_layout()


plt.savefig(
    COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    COMPARISON_GRAPH_PATH
)


# =============================================================================
# 36. HEATMAP
# =============================================================================

print("\n" + "=" * 80)
print("CREATING CORRELATION HEATMAP")
print("=" * 80)


numeric_df = train_df.select_dtypes(
    include=np.number
).copy()


numeric_df = numeric_df.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


numeric_df = numeric_df.dropna(
    axis=1,
    how="all"
)


# =============================================================================
# Find variables most correlated with target
# =============================================================================

if TARGET_COLUMN in numeric_df.columns:

    target_correlations = (
        numeric_df
        .corr()[
            TARGET_COLUMN
        ]
        .abs()
        .sort_values(
            ascending=False
        )
    )


    selected_columns = (
        target_correlations
        .head(12)
        .index
        .tolist()
    )


else:

    selected_columns = (
        numeric_df
        .columns[:12]
        .tolist()
    )


print(
    "\nVariables used in heatmap:"
)


for column in selected_columns:

    print(
        " -",
        column
    )


correlation_matrix = numeric_df[
    selected_columns
].corr()


# =============================================================================
# DRAW HEATMAP
# =============================================================================

fig, ax = plt.subplots(
    figsize=(
        14,
        11
    )
)


image = ax.imshow(
    correlation_matrix.values,
    vmin=-1,
    vmax=1,
    aspect="auto"
)


colorbar = fig.colorbar(
    image,
    ax=ax
)


colorbar.set_label(
    "Correlation"
)


ax.set_xticks(
    np.arange(
        len(
            selected_columns
        )
    )
)


ax.set_yticks(
    np.arange(
        len(
            selected_columns
        )
    )
)


ax.set_xticklabels(
    selected_columns,
    rotation=90,
    fontsize=8
)


ax.set_yticklabels(
    selected_columns,
    fontsize=8
)


# =============================================================================
# Correlation values
# =============================================================================

for row in range(
    len(
        selected_columns
    )
):

    for column in range(
        len(
            selected_columns
        )
    ):

        value = correlation_matrix.iloc[
            row,
            column
        ]


        if not np.isnan(
            value
        ):

            ax.text(
                column,
                row,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=7
            )


ax.set_title(
    "Correlation Heatmap - Farmer Income Dataset",
    pad=20
)


plt.tight_layout()


plt.savefig(
    HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    HEATMAP_PATH
)


# =============================================================================
# 37. RESULT CSV
# =============================================================================

print("\n" + "=" * 80)
print("CREATING RESULT CSV")
print("=" * 80)


result_df = pd.DataFrame(
    {
        "Actual_Income":
            y_valid_array,

        "Predicted_Income":
            valid_predictions
    }
)


result_df[
    "Residual"
] = (
    result_df[
        "Actual_Income"
    ]
    -
    result_df[
        "Predicted_Income"
    ]
)


result_df[
    "Absolute_Error"
] = np.abs(
    result_df[
        "Residual"
    ]
)


actual_absolute = np.abs(
    result_df[
        "Actual_Income"
    ].values
)


result_df[
    "Percentage_Error"
] = np.where(

    actual_absolute > 1e-8,

    (
        result_df[
            "Absolute_Error"
        ].values
        /
        actual_absolute
    )
    * 100,

    np.nan
)


result_df.to_csv(
    RESULT_CSV_PATH,
    index=False
)


print(
    "\nSaved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nFirst 10 results:"
)


print(
    result_df.head(
        10
    ).to_string(
        index=False
    )
)


# =============================================================================
# 38. RESULT GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING RESULT GRAPH")
print("=" * 80)


plt.figure(
    figsize=(
        10,
        7
    )
)


plt.scatter(

    result_df[
        "Actual_Income"
    ],

    result_df[
        "Predicted_Income"
    ],

    alpha=0.45,

    s=20
)


minimum_value = min(

    result_df[
        "Actual_Income"
    ].min(),

    result_df[
        "Predicted_Income"
    ].min()
)


maximum_value = max(

    result_df[
        "Actual_Income"
    ].max(),

    result_df[
        "Predicted_Income"
    ].max()
)


plt.plot(

    [
        minimum_value,
        maximum_value
    ],

    [
        minimum_value,
        maximum_value
    ],

    linestyle="--",

    linewidth=2,

    label="Perfect Prediction"
)


plt.xlabel(
    "Actual Farmer Income"
)


plt.ylabel(
    "Predicted Farmer Income"
)


plt.title(
    "Actual vs Predicted Farmer Income"
)


plt.legend()


plt.grid(
    alpha=0.25
)


text = (

    f"R² = {valid_r2:.4f}\n"

    f"MAE = {valid_mae:,.2f}\n"

    f"RMSE = {valid_rmse:,.2f}"
)


plt.text(

    0.05,

    0.95,

    text,

    transform=plt.gca().transAxes,

    verticalalignment="top",

    bbox=dict(
        boxstyle="round",
        alpha=0.2
    )
)


plt.tight_layout()


plt.savefig(
    RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    RESULT_GRAPH_PATH
)


# =============================================================================
# 39. PREDICTION CSV
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PREDICTION CSV")
print("=" * 80)


prediction_df = test_df.copy()


if TARGET_COLUMN in prediction_df.columns:

    prediction_df = prediction_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    )


prediction_df[
    "Predicted_Total_Income"
] = test_predictions


prediction_df.insert(

    0,

    "Prediction_ID",

    np.arange(
        1,
        len(
            prediction_df
        )
        + 1
    )
)


prediction_df.to_csv(
    PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nSaved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nFirst 10 predictions:"
)


print(

    prediction_df[
        [
            "Prediction_ID",
            "Predicted_Total_Income"
        ]
    ]
    .head(
        10
    )
    .to_string(
        index=False
    )
)


# =============================================================================
# 40. PREDICTION GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PREDICTION GRAPH")
print("=" * 80)


sorted_predictions = np.sort(
    test_predictions
)


record_numbers = np.arange(
    1,
    len(
        sorted_predictions
    )
    + 1
)


mean_prediction = np.mean(
    test_predictions
)


plt.figure(
    figsize=(
        12,
        6
    )
)


plt.plot(
    record_numbers,
    sorted_predictions,
    linewidth=2,
    label="Predicted Income"
)


plt.axhline(
    mean_prediction,
    linestyle="--",
    linewidth=2,
    label=(
        f"Mean = "
        f"{mean_prediction:,.2f}"
    )
)


plt.xlabel(
    "Test Records Sorted by Predicted Income"
)


plt.ylabel(
    "Predicted Farmer Income"
)


plt.title(
    "Farmer Income Prediction Distribution"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    PREDICTION_GRAPH_PATH
)


# =============================================================================
# 41. ADDITIONAL RESIDUAL GRAPH
# =============================================================================

RESIDUAL_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "residual_graph.png"
)


print("\n" + "=" * 80)
print("CREATING RESIDUAL GRAPH")
print("=" * 80)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.scatter(

    result_df[
        "Predicted_Income"
    ],

    result_df[
        "Residual"
    ],

    alpha=0.45,

    s=20
)


plt.axhline(
    0,
    linestyle="--",
    linewidth=2
)


plt.xlabel(
    "Predicted Farmer Income"
)


plt.ylabel(
    "Residual (Actual - Predicted)"
)


plt.title(
    "Residual Analysis - Farmer Income Prediction"
)


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    RESIDUAL_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    RESIDUAL_GRAPH_PATH
)


# =============================================================================
# 42. ACTUAL VS PREDICTED SEQUENCE GRAPH
# =============================================================================

SEQUENCE_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "actual_predicted_comparison_graph.png"
)


print("\n" + "=" * 80)
print("CREATING ACTUAL VS PREDICTED COMPARISON")
print("=" * 80)


# Sort by actual income for cleaner graph

comparison_data = result_df.sort_values(
    by="Actual_Income"
).reset_index(
    drop=True
)


# If dataset is huge, plotting every record can be visually dense.
# We still plot all records as requested.

indices = np.arange(
    len(
        comparison_data
    )
)


plt.figure(
    figsize=(
        14,
        7
    )
)


plt.plot(

    indices,

    comparison_data[
        "Actual_Income"
    ].values,

    label="Actual Income",

    linewidth=2
)


plt.plot(

    indices,

    comparison_data[
        "Predicted_Income"
    ].values,

    label="Predicted Income",

    linewidth=2
)


plt.xlabel(
    "Validation Records Sorted by Actual Income"
)


plt.ylabel(
    "Farmer Income"
)


plt.title(
    "Actual vs Predicted Farmer Income Comparison"
)


plt.legend()


plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    SEQUENCE_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


plt.close()


print(
    "Saved:",
    SEQUENCE_GRAPH_PATH
)


# =============================================================================
# 43. PREDICTION SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("TEST PREDICTION SUMMARY")
print("=" * 80)


print(
    f"\nNumber of predictions : "
    f"{len(test_predictions):,}"
)


print(
    f"Minimum income        : "
    f"{np.min(test_predictions):,.2f}"
)


print(
    f"Maximum income        : "
    f"{np.max(test_predictions):,.2f}"
)


print(
    f"Mean income           : "
    f"{np.mean(test_predictions):,.2f}"
)


print(
    f"Median income         : "
    f"{np.median(test_predictions):,.2f}"
)


print(
    f"Standard deviation    : "
    f"{np.std(test_predictions):,.2f}"
)


# =============================================================================
# 44. VERIFY FILES
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING GENERATED FILES")
print("=" * 80)


generated_files = [

    ACCURACY_GRAPH_PATH,

    COMPARISON_GRAPH_PATH,

    HEATMAP_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    METRICS_CSV_PATH,

    RESIDUAL_GRAPH_PATH,

    SEQUENCE_GRAPH_PATH
]


for path in generated_files:

    if os.path.isfile(
        path
    ):

        size_kb = (
            os.path.getsize(
                path
            )
            /
            1024
        )


        print(
            f"\n[OK] "
            f"{os.path.basename(path)}"
        )


        print(
            f"     Size: "
            f"{size_kb:,.2f} KB"
        )


    else:

        print(
            f"\n[ERROR] "
            f"{os.path.basename(path)}"
        )


# =============================================================================
# 45. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 80)


print(
    "\nOutput directory:"
)


print(
    PROJECT_DIR
)


print(
    "\nGenerated files:"
)


print(
    """
1.  accuracy_graph.png
2.  comparison_graph.png
3.  heatmap.png
4.  result.csv
5.  result_graph.png
6.  prediction.csv
7.  prediction_graph.png
8.  model_metrics.csv
9.  residual_graph.png
10. actual_predicted_comparison_graph.png
"""
)


print(
    "FINAL VALIDATION PERFORMANCE"
)


print(
    f"\nR²   = "
    f"{valid_r2:.6f}"
)


print(
    f"MAE  = "
    f"{valid_mae:,.4f}"
)


print(
    f"MSE  = "
    f"{valid_mse:,.4f}"
)


print(
    f"RMSE = "
    f"{valid_rmse:,.4f}"
)


print(
    f"Regression Score = "
    f"{valid_score * 100:.2f}%"
)


print("\n" + "=" * 80)
print("DONE")
print("=" * 80)
```
