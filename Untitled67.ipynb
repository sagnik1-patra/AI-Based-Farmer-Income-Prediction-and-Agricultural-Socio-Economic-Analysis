```python
# =============================================================================
# AI-BASED FARMER INCOME PREDICTION AND
# AGRICULTURAL SOCIO-ECONOMIC ANALYSIS
#
# FIXED VERSION:
# - Handles real ZIP file
# - Handles folder named farmer-income-data.zip
# - Handles PermissionError while opening ZIP
# - Copies blocked ZIP to temporary location when possible
# - Automatically locates CSV files
# - Saves H5, PKL, YAML and JSON
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import json
import yaml
import pickle
import zipfile
import shutil
import tempfile
import warnings

import numpy as np
import pandas as pd

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Dense,
    Dropout,
    BatchNormalization,
    Input
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# =============================================================================
# 2. SETTINGS
# =============================================================================

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =============================================================================
# 3. PROJECT PATHS
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Farmer Income Prediction and Agricultural Socio-Economic Analysis"
)

DATA_SOURCE = os.path.join(
    PROJECT_DIR,
    "farmer-income-data.zip"
)


TRAIN_FILE = "lte_train.csv"
TEST_FILE = "lte_test.csv"
DICTIONARY_FILE = "lte_dictionary.csv"


# =============================================================================
# 4. OUTPUT PATHS
# =============================================================================

H5_PATH = os.path.join(
    PROJECT_DIR,
    "farmer_income_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "farmer_income_preprocessor.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "farmer_income_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "farmer_income_metadata.json"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 5. PROJECT INFORMATION
# =============================================================================

print("=" * 80)
print("AI-BASED FARMER INCOME PREDICTION")
print("=" * 80)

print("\nProject directory:")
print(PROJECT_DIR)

print("\nExpected data source:")
print(DATA_SOURCE)


# =============================================================================
# 6. CHECK DATA SOURCE
# =============================================================================

print("\n" + "=" * 80)
print("CHECKING DATA SOURCE")
print("=" * 80)


if not os.path.exists(DATA_SOURCE):

    print("\nThe expected ZIP/folder was not found.")

    print("\nSearching project directory for CSV files...")

    csv_files = []

    for root, dirs, files in os.walk(PROJECT_DIR):

        for file in files:

            if file.lower().endswith(".csv"):

                csv_files.append(
                    os.path.join(root, file)
                )

    if len(csv_files) == 0:

        raise FileNotFoundError(
            "\nCould not find farmer-income-data.zip "
            "or any CSV files in:\n"
            f"{PROJECT_DIR}"
        )

    else:

        print("\nCSV files found:")

        for file in csv_files:
            print(file)


else:

    print("\nData source exists.")

    if os.path.isdir(DATA_SOURCE):

        print(
            "Detected type: DIRECTORY "
            "(folder named farmer-income-data.zip)"
        )

    elif os.path.isfile(DATA_SOURCE):

        print(
            "Detected type: FILE"
        )


# =============================================================================
# 7. HELPER FUNCTION - FIND FILE RECURSIVELY
# =============================================================================

def find_file_in_directory(directory, required_name):

    required_name = required_name.lower()

    for root, dirs, files in os.walk(directory):

        for file in files:

            if file.lower() == required_name:

                return os.path.join(
                    root,
                    file
                )

    return None


# =============================================================================
# 8. HELPER FUNCTION - FIND FILE INSIDE ZIP
# =============================================================================

def find_file_in_zip(zip_names, required_name):

    required_name = required_name.lower()

    # First try exact basename match

    for name in zip_names:

        if (
            os.path.basename(name).lower()
            == required_name
        ):

            return name

    return None


# =============================================================================
# 9. FUNCTION TO SAFELY OPEN ZIP
# =============================================================================

def get_safe_zip_path(original_zip):

    print("\nChecking ZIP accessibility...")

    # -------------------------------------------------------------------------
    # Try opening original ZIP
    # -------------------------------------------------------------------------

    try:

        with open(original_zip, "rb") as f:

            f.read(4)

        print(
            "ZIP file can be accessed directly."
        )

        return original_zip


    except PermissionError:

        print(
            "\nPermissionError detected while accessing ZIP."
        )

        print(
            "Attempting to copy ZIP to a temporary location..."
        )


    except Exception as e:

        print(
            "\nDirect ZIP access produced:"
        )

        print(type(e).__name__, str(e))

        print(
            "\nAttempting temporary copy..."
        )


    # -------------------------------------------------------------------------
    # Copy ZIP to temp directory
    # -------------------------------------------------------------------------

    temp_dir = tempfile.mkdtemp(
        prefix="farmer_income_"
    )

    temp_zip = os.path.join(
        temp_dir,
        "farmer-income-data.zip"
    )


    try:

        shutil.copy2(
            original_zip,
            temp_zip
        )

        print(
            "\nZIP copied successfully to:"
        )

        print(temp_zip)

        return temp_zip


    except PermissionError as e:

        raise PermissionError(
            "\nWindows is completely blocking access to:\n"
            f"{original_zip}\n\n"
            "Please close File Explorer preview, WinRAR, "
            "7-Zip or any application using the ZIP.\n\n"
            "If the problem continues, extract the ZIP manually "
            "into the project directory. This code will automatically "
            "detect the extracted CSV files."
        ) from e


# =============================================================================
# 10. LOAD DATASETS
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATASETS")
print("=" * 80)


train_df = None
test_df = None
dictionary_df = None


# =============================================================================
# CASE 1:
# farmer-income-data.zip IS ACTUALLY A DIRECTORY
# =============================================================================

if os.path.isdir(DATA_SOURCE):

    print(
        "\nfarmer-income-data.zip is actually a directory."
    )

    print(
        "Reading CSV files directly from directory..."
    )


    train_path = find_file_in_directory(
        DATA_SOURCE,
        TRAIN_FILE
    )

    test_path = find_file_in_directory(
        DATA_SOURCE,
        TEST_FILE
    )

    dictionary_path = find_file_in_directory(
        DATA_SOURCE,
        DICTIONARY_FILE
    )


    if train_path is None:

        raise FileNotFoundError(
            f"{TRAIN_FILE} not found inside {DATA_SOURCE}"
        )


    if test_path is None:

        raise FileNotFoundError(
            f"{TEST_FILE} not found inside {DATA_SOURCE}"
        )


    if dictionary_path is None:

        raise FileNotFoundError(
            f"{DICTIONARY_FILE} not found inside {DATA_SOURCE}"
        )


    print("\nTrain:")
    print(train_path)

    print("\nTest:")
    print(test_path)

    print("\nDictionary:")
    print(dictionary_path)


    train_df = pd.read_csv(
        train_path,
        low_memory=False
    )

    test_df = pd.read_csv(
        test_path,
        low_memory=False
    )

    dictionary_df = pd.read_csv(
        dictionary_path,
        low_memory=False
    )


# =============================================================================
# CASE 2:
# DATA SOURCE IS A REAL ZIP FILE
# =============================================================================

elif os.path.isfile(DATA_SOURCE):

    print(
        "\nfarmer-income-data.zip detected as a file."
    )


    safe_zip_path = get_safe_zip_path(
        DATA_SOURCE
    )


    # -------------------------------------------------------------------------
    # Check if it is actually a ZIP
    # -------------------------------------------------------------------------

    try:

        is_zip = zipfile.is_zipfile(
            safe_zip_path
        )

    except PermissionError:

        is_zip = False


    if not is_zip:

        raise ValueError(
            "\nThe file exists but Python does not recognize it "
            "as a valid ZIP archive:\n"
            f"{DATA_SOURCE}\n\n"
            "Please extract it manually and place the CSV files "
            "inside the project folder."
        )


    print("\nValid ZIP archive detected.")


    # -------------------------------------------------------------------------
    # Read ZIP
    # -------------------------------------------------------------------------

    with zipfile.ZipFile(
        safe_zip_path,
        "r"
    ) as zip_ref:

        zip_names = zip_ref.namelist()


        print("\nFiles inside ZIP:")

        for name in zip_names:
            print(" -", name)


        train_zip_name = find_file_in_zip(
            zip_names,
            TRAIN_FILE
        )

        test_zip_name = find_file_in_zip(
            zip_names,
            TEST_FILE
        )

        dictionary_zip_name = find_file_in_zip(
            zip_names,
            DICTIONARY_FILE
        )


        if train_zip_name is None:

            raise FileNotFoundError(
                f"{TRAIN_FILE} not found inside ZIP."
            )


        if test_zip_name is None:

            raise FileNotFoundError(
                f"{TEST_FILE} not found inside ZIP."
            )


        if dictionary_zip_name is None:

            raise FileNotFoundError(
                f"{DICTIONARY_FILE} not found inside ZIP."
            )


        print("\nLocated files:")

        print(
            "Train:",
            train_zip_name
        )

        print(
            "Test:",
            test_zip_name
        )

        print(
            "Dictionary:",
            dictionary_zip_name
        )


        with zip_ref.open(
            train_zip_name
        ) as file:

            train_df = pd.read_csv(
                file,
                low_memory=False
            )


        with zip_ref.open(
            test_zip_name
        ) as file:

            test_df = pd.read_csv(
                file,
                low_memory=False
            )


        with zip_ref.open(
            dictionary_zip_name
        ) as file:

            dictionary_df = pd.read_csv(
                file,
                low_memory=False
            )


# =============================================================================
# CASE 3:
# ZIP NOT FOUND - SEARCH PROJECT DIRECTORY
# =============================================================================

else:

    print(
        "\nZIP not found. Searching project directory..."
    )


    train_path = find_file_in_directory(
        PROJECT_DIR,
        TRAIN_FILE
    )

    test_path = find_file_in_directory(
        PROJECT_DIR,
        TEST_FILE
    )

    dictionary_path = find_file_in_directory(
        PROJECT_DIR,
        DICTIONARY_FILE
    )


    if train_path is None:

        raise FileNotFoundError(
            f"\nCould not find {TRAIN_FILE}"
        )


    if test_path is None:

        raise FileNotFoundError(
            f"\nCould not find {TEST_FILE}"
        )


    if dictionary_path is None:

        raise FileNotFoundError(
            f"\nCould not find {DICTIONARY_FILE}"
        )


    print("\nUsing extracted files:")

    print(train_path)
    print(test_path)
    print(dictionary_path)


    train_df = pd.read_csv(
        train_path,
        low_memory=False
    )

    test_df = pd.read_csv(
        test_path,
        low_memory=False
    )

    dictionary_df = pd.read_csv(
        dictionary_path,
        low_memory=False
    )


# =============================================================================
# 11. CONFIRM DATA LOADING
# =============================================================================

print("\n" + "=" * 80)
print("DATASETS LOADED SUCCESSFULLY")
print("=" * 80)


print("\nTraining dataset:")
print(train_df.shape)


print("\nTest dataset:")
print(test_df.shape)


print("\nData dictionary:")
print(dictionary_df.shape)


print("\nTraining data preview:")

print(
    train_df.head()
)


print("\nDictionary preview:")

print(
    dictionary_df.head(10).to_string()
)


# =============================================================================
# 12. CLEAN COLUMN NAMES
# =============================================================================

train_df.columns = [
    str(col).strip()
    for col in train_df.columns
]

test_df.columns = [
    str(col).strip()
    for col in test_df.columns
]

dictionary_df.columns = [
    str(col).strip()
    for col in dictionary_df.columns
]


# =============================================================================
# 13. PRINT ALL TRAINING COLUMNS
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING COLUMNS")
print("=" * 80)


for i, column in enumerate(
    train_df.columns,
    start=1
):

    print(
        f"{i:3d}. {column}"
    )


# =============================================================================
# 14. TARGET DETECTION
# =============================================================================

print("\n" + "=" * 80)
print("DETECTING TARGET VARIABLE")
print("=" * 80)


TARGET_COLUMN = None


preferred_targets = [

    "Target_Variable/Total Income",

    "Target_Variable/Total income",

    "Total Income",

    "Total income",

    "Total_Income",

    "total_income"
]


# =============================================================================
# Exact match
# =============================================================================

for candidate in preferred_targets:

    if candidate in train_df.columns:

        TARGET_COLUMN = candidate

        break


# =============================================================================
# Case-insensitive match
# =============================================================================

if TARGET_COLUMN is None:

    column_lookup = {

        str(column).lower().strip():
            column

        for column in train_df.columns
    }


    for candidate in preferred_targets:

        key = candidate.lower().strip()

        if key in column_lookup:

            TARGET_COLUMN = column_lookup[key]

            break


# =============================================================================
# Search for Total + Income
# =============================================================================

if TARGET_COLUMN is None:

    income_columns = [

        column

        for column in train_df.columns

        if "income" in str(column).lower()
    ]


    print("\nIncome-related columns:")

    for column in income_columns:

        print(
            " -",
            column
        )


    for column in income_columns:

        name = str(
            column
        ).lower()

        if (
            "total" in name
            and
            "income" in name
        ):

            TARGET_COLUMN = column

            break


# =============================================================================
# Final target check
# =============================================================================

if TARGET_COLUMN is None:

    raise ValueError(
        "\nTarget variable could not be automatically identified.\n"
        "Check the list of columns printed above."
    )


print("\nTarget detected:")

print(
    TARGET_COLUMN
)


# =============================================================================
# 15. CLEAN TARGET
# =============================================================================

print("\n" + "=" * 80)
print("CLEANING TARGET")
print("=" * 80)


train_df[TARGET_COLUMN] = pd.to_numeric(
    train_df[TARGET_COLUMN],
    errors="coerce"
)


train_df[TARGET_COLUMN] = train_df[
    TARGET_COLUMN
].replace(
    [np.inf, -np.inf],
    np.nan
)


missing_target = train_df[
    TARGET_COLUMN
].isna().sum()


print(
    "\nMissing target values:",
    missing_target
)


train_df = train_df.dropna(
    subset=[TARGET_COLUMN]
).reset_index(
    drop=True
)


print(
    "\nRows after target cleaning:",
    len(train_df)
)


print("\nTarget statistics:")

print(
    train_df[TARGET_COLUMN].describe()
)


# =============================================================================
# 16. CREATE X AND Y
# =============================================================================

X = train_df.drop(
    columns=[
        TARGET_COLUMN
    ]
).copy()


y = train_df[
    TARGET_COLUMN
].astype(
    np.float64
).copy()


if TARGET_COLUMN in test_df.columns:

    X_external_test = test_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    ).copy()

else:

    X_external_test = test_df.copy()


print(
    "\nOriginal feature count:",
    X.shape[1]
)


# =============================================================================
# 17. REPLACE INFINITE VALUES
# =============================================================================

X = X.replace(
    [np.inf, -np.inf],
    np.nan
)


X_external_test = X_external_test.replace(
    [np.inf, -np.inf],
    np.nan
)


# =============================================================================
# 18. REMOVE COMPLETELY EMPTY FEATURES
# =============================================================================

empty_columns = [

    column

    for column in X.columns

    if X[column].isna().all()
]


print(
    "\nCompletely empty columns:",
    len(empty_columns)
)


if empty_columns:

    X.drop(
        columns=empty_columns,
        inplace=True
    )

    X_external_test.drop(
        columns=[
            column
            for column in empty_columns
            if column in X_external_test.columns
        ],
        inplace=True,
        errors="ignore"
    )


# =============================================================================
# 19. REMOVE CONSTANT FEATURES
# =============================================================================

constant_columns = [

    column

    for column in X.columns

    if X[column].nunique(
        dropna=False
    ) <= 1
]


print(
    "Constant columns:",
    len(constant_columns)
)


if constant_columns:

    X.drop(
        columns=constant_columns,
        inplace=True
    )

    X_external_test.drop(
        columns=[
            column
            for column in constant_columns
            if column in X_external_test.columns
        ],
        inplace=True,
        errors="ignore"
    )


# =============================================================================
# 20. ALIGN TEST COLUMNS
# =============================================================================

for column in X.columns:

    if column not in X_external_test.columns:

        X_external_test[column] = np.nan


extra_test_columns = [

    column

    for column in X_external_test.columns

    if column not in X.columns
]


if extra_test_columns:

    X_external_test.drop(
        columns=extra_test_columns,
        inplace=True
    )


X_external_test = X_external_test[
    X.columns
]


print(
    "\nFinal training feature shape:",
    X.shape
)

print(
    "Final external test shape:",
    X_external_test.shape
)


# =============================================================================
# 21. IDENTIFY NUMERIC AND CATEGORICAL FEATURES
# =============================================================================

numeric_features = X.select_dtypes(
    include=[
        np.number,
        "bool"
    ]
).columns.tolist()


categorical_features = X.select_dtypes(
    exclude=[
        np.number,
        "bool"
    ]
).columns.tolist()


print("\n" + "=" * 80)
print("FEATURE TYPES")
print("=" * 80)


print(
    "\nNumeric features:",
    len(numeric_features)
)


print(
    "Categorical features:",
    len(categorical_features)
)


# =============================================================================
# 22. NUMERIC PIPELINE
# =============================================================================

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",

            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",

            StandardScaler()
        )
    ]
)


# =============================================================================
# 23. CATEGORICAL PIPELINE
# =============================================================================

try:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )

except TypeError:

    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",

            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",

            encoder
        )
    ]
)


# =============================================================================
# 24. COLUMN TRANSFORMER
# =============================================================================

transformers = []


if numeric_features:

    transformers.append(
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        )
    )


if categorical_features:

    transformers.append(
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    )


preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)


# =============================================================================
# 25. TRAIN VALIDATION SPLIT
# =============================================================================

print("\n" + "=" * 80)
print("TRAIN / VALIDATION SPLIT")
print("=" * 80)


X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print(
    "\nTraining rows:",
    len(X_train)
)

print(
    "Validation rows:",
    len(X_valid)
)


# =============================================================================
# 26. FIT PREPROCESSOR
# =============================================================================

print("\n" + "=" * 80)
print("PREPROCESSING")
print("=" * 80)


X_train_processed = preprocessor.fit_transform(
    X_train
)


X_valid_processed = preprocessor.transform(
    X_valid
)


X_test_processed = preprocessor.transform(
    X_external_test
)


X_train_processed = np.asarray(
    X_train_processed,
    dtype=np.float32
)


X_valid_processed = np.asarray(
    X_valid_processed,
    dtype=np.float32
)


X_test_processed = np.asarray(
    X_test_processed,
    dtype=np.float32
)


print(
    "\nProcessed training shape:",
    X_train_processed.shape
)


print(
    "Processed validation shape:",
    X_valid_processed.shape
)


print(
    "Processed test shape:",
    X_test_processed.shape
)


# =============================================================================
# 27. CHECK FOR INVALID VALUES
# =============================================================================

print("\nChecking processed data...")


print(
    "Training NaN:",
    np.isnan(
        X_train_processed
    ).sum()
)


print(
    "Training Inf:",
    np.isinf(
        X_train_processed
    ).sum()
)


# =============================================================================
# 28. TARGET SCALING
# =============================================================================

target_scaler = StandardScaler()


y_train_scaled = target_scaler.fit_transform(
    np.asarray(
        y_train
    ).reshape(
        -1,
        1
    )
).flatten()


y_valid_scaled = target_scaler.transform(
    np.asarray(
        y_valid
    ).reshape(
        -1,
        1
    )
).flatten()


# =============================================================================
# 29. BUILD ANN
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING ANN MODEL")
print("=" * 80)


INPUT_DIM = X_train_processed.shape[1]


model = Sequential(
    [

        Input(
            shape=(
                INPUT_DIM,
            )
        ),

        Dense(
            256,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.30
        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.25
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )
    ]
)


# =============================================================================
# 30. COMPILE MODEL
# =============================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=[
        "mae"
    ]
)


model.summary()


# =============================================================================
# 31. CALLBACKS
# =============================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=15,

    restore_best_weights=True,

    verbose=1
)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=7,

    min_lr=1e-6,

    verbose=1
)


# =============================================================================
# 32. TRAIN MODEL
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING MODEL")
print("=" * 80)


history = model.fit(

    X_train_processed,

    y_train_scaled,

    validation_data=(
        X_valid_processed,
        y_valid_scaled
    ),

    epochs=150,

    batch_size=64,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)


# =============================================================================
# 33. VALIDATION PREDICTIONS
# =============================================================================

print("\n" + "=" * 80)
print("VALIDATION PREDICTIONS")
print("=" * 80)


valid_prediction_scaled = model.predict(
    X_valid_processed,
    verbose=0
)


valid_predictions = target_scaler.inverse_transform(
    valid_prediction_scaled
).flatten()


# =============================================================================
# 34. PERFORMANCE METRICS
# =============================================================================

mae = mean_absolute_error(
    y_valid,
    valid_predictions
)


mse = mean_squared_error(
    y_valid,
    valid_predictions
)


rmse = np.sqrt(
    mse
)


r2 = r2_score(
    y_valid,
    valid_predictions
)


print(
    f"\nMAE  : {mae:,.4f}"
)

print(
    f"MSE  : {mse:,.4f}"
)

print(
    f"RMSE : {rmse:,.4f}"
)

print(
    f"R²   : {r2:.6f}"
)


# =============================================================================
# 35. EXTERNAL TEST PREDICTIONS
# =============================================================================

print("\n" + "=" * 80)
print("TEST DATA PREDICTIONS")
print("=" * 80)


test_prediction_scaled = model.predict(
    X_test_processed,
    verbose=0
)


test_predictions = target_scaler.inverse_transform(
    test_prediction_scaled
).flatten()


print(
    "\nPredictions generated:",
    len(test_predictions)
)


print("\nFirst 10 predictions:")


for index, prediction in enumerate(
    test_predictions[:10],
    start=1
):

    print(
        f"{index:2d}. "
        f"{prediction:,.2f}"
    )


# =============================================================================
# 36. SAVE H5
# =============================================================================

print("\n" + "=" * 80)
print("SAVING H5")
print("=" * 80)


model.save(
    H5_PATH
)


print(
    "\nSaved:"
)

print(
    H5_PATH
)


# =============================================================================
# 37. SAVE PKL
# =============================================================================

print("\n" + "=" * 80)
print("SAVING PKL")
print("=" * 80)


preprocessor_package = {

    "preprocessor":
        preprocessor,

    "target_scaler":
        target_scaler,

    "target_column":
        TARGET_COLUMN,

    "feature_columns":
        X.columns.tolist(),

    "numeric_features":
        numeric_features,

    "categorical_features":
        categorical_features,

    "empty_columns_removed":
        empty_columns,

    "constant_columns_removed":
        constant_columns
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        preprocessor_package,
        file
    )


print(
    "\nSaved:"
)

print(
    PKL_PATH
)


# =============================================================================
# 38. YAML CONFIGURATION
# =============================================================================

print("\n" + "=" * 80)
print("SAVING YAML")
print("=" * 80)


yaml_data = {

    "project": {

        "name":
            "AI-Based Farmer Income Prediction and "
            "Agricultural Socio-Economic Analysis",

        "problem_type":
            "Regression",

        "target":
            str(TARGET_COLUMN)
    },


    "dataset": {

        "training_rows":
            int(
                train_df.shape[0]
            ),

        "test_rows":
            int(
                test_df.shape[0]
            ),

        "feature_count":
            int(
                X.shape[1]
            ),

        "numeric_features":
            int(
                len(
                    numeric_features
                )
            ),

        "categorical_features":
            int(
                len(
                    categorical_features
                )
            ),

        "processed_features":
            int(
                X_train_processed.shape[1]
            )
    },


    "preprocessing": {

        "numeric_missing":
            "median",

        "categorical_missing":
            "most_frequent",

        "numeric_scaler":
            "StandardScaler",

        "categorical_encoder":
            "OneHotEncoder",

        "target_scaler":
            "StandardScaler"
    },


    "model": {

        "type":
            "Artificial Neural Network",

        "layers":
            [
                256,
                128,
                64,
                32
            ],

        "optimizer":
            "Adam",

        "learning_rate":
            0.001,

        "loss":
            "MSE",

        "epochs":
            150,

        "batch_size":
            64
    },


    "validation_metrics": {

        "mae":
            float(
                mae
            ),

        "mse":
            float(
                mse
            ),

        "rmse":
            float(
                rmse
            ),

        "r2":
            float(
                r2
            )
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nSaved:"
)

print(
    YAML_PATH
)


# =============================================================================
# 39. JSON METADATA
# =============================================================================

print("\n" + "=" * 80)
print("SAVING JSON")
print("=" * 80)


json_data = {

    "project":
        "AI-Based Farmer Income Prediction and "
        "Agricultural Socio-Economic Analysis",

    "problem_type":
        "Regression",

    "target":
        str(
            TARGET_COLUMN
        ),

    "dataset": {

        "train_rows":
            int(
                train_df.shape[0]
            ),

        "train_columns":
            int(
                train_df.shape[1]
            ),

        "test_rows":
            int(
                test_df.shape[0]
            ),

        "test_columns":
            int(
                test_df.shape[1]
            ),

        "dictionary_rows":
            int(
                dictionary_df.shape[0]
            )
    },

    "features": {

        "feature_count":
            int(
                X.shape[1]
            ),

        "numeric_count":
            int(
                len(
                    numeric_features
                )
            ),

        "categorical_count":
            int(
                len(
                    categorical_features
                )
            ),

        "processed_count":
            int(
                X_train_processed.shape[1]
            ),

        "numeric_features":
            [
                str(x)
                for x in numeric_features
            ],

        "categorical_features":
            [
                str(x)
                for x in categorical_features
            ]
    },

    "model": {

        "framework":
            "TensorFlow/Keras",

        "type":
            "Artificial Neural Network",

        "input_dimension":
            int(
                INPUT_DIM
            ),

        "hidden_layers":
            [
                256,
                128,
                64,
                32
            ]
    },

    "validation_metrics": {

        "MAE":
            float(
                mae
            ),

        "MSE":
            float(
                mse
            ),

        "RMSE":
            float(
                rmse
            ),

        "R2":
            float(
                r2
            )
    },

    "target_statistics": {

        "minimum":
            float(
                y.min()
            ),

        "maximum":
            float(
                y.max()
            ),

        "mean":
            float(
                y.mean()
            ),

        "median":
            float(
                y.median()
            ),

        "std":
            float(
                y.std()
            )
    },

    "test_prediction_statistics": {

        "count":
            int(
                len(
                    test_predictions
                )
            ),

        "minimum":
            float(
                np.min(
                    test_predictions
                )
            ),

        "maximum":
            float(
                np.max(
                    test_predictions
                )
            ),

        "mean":
            float(
                np.mean(
                    test_predictions
                )
            )
    },

    "files": {

        "h5":
            H5_PATH,

        "pkl":
            PKL_PATH,

        "yaml":
            YAML_PATH,

        "json":
            JSON_PATH
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4,
        ensure_ascii=False
    )


print(
    "\nSaved:"
)

print(
    JSON_PATH
)


# =============================================================================
# 40. VERIFY OUTPUT FILES
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING GENERATED FILES")
print("=" * 80)


output_files = [

    H5_PATH,

    PKL_PATH,

    YAML_PATH,

    JSON_PATH
]


for path in output_files:

    if os.path.isfile(path):

        size_mb = (
            os.path.getsize(
                path
            )
            /
            1024
            /
            1024
        )

        print(
            f"\n[OK] "
            f"{os.path.basename(path)}"
        )

        print(
            f"     {size_mb:.3f} MB"
        )

    else:

        print(
            "\n[ERROR]",
            path
        )


# =============================================================================
# 41. VERIFY PKL
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING PKL")
print("=" * 80)


with open(
    PKL_PATH,
    "rb"
) as file:

    loaded_package = pickle.load(
        file
    )


print(
    "\nPKL loaded successfully."
)


print(
    "Target:",
    loaded_package[
        "target_column"
    ]
)


# =============================================================================
# 42. VERIFY H5
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING H5")
print("=" * 80)


loaded_model = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)


print(
    "\nH5 model loaded successfully."
)


sample_scaled_prediction = loaded_model.predict(
    X_valid_processed[:1],
    verbose=0
)


sample_prediction = loaded_package[
    "target_scaler"
].inverse_transform(
    sample_scaled_prediction
).flatten()[0]


sample_actual = float(
    y_valid.iloc[0]
)


print(
    f"\nActual income:"
    f"    {sample_actual:,.2f}"
)


print(
    f"Predicted income:"
    f" {sample_prediction:,.2f}"
)


# =============================================================================
# 43. FINAL RESULTS
# =============================================================================

print("\n" + "=" * 80)
print("FINAL RESULTS")
print("=" * 80)


print(
    "\nTarget:",
    TARGET_COLUMN
)


print(
    "\nTraining records:",
    f"{len(train_df):,}"
)


print(
    "Test records:",
    f"{len(test_df):,}"
)


print(
    "Original features:",
    X.shape[1]
)


print(
    "ANN input features:",
    X_train_processed.shape[1]
)


print("\nMODEL PERFORMANCE")


print(
    f"MAE  = {mae:,.4f}"
)


print(
    f"RMSE = {rmse:,.4f}"
)


print(
    f"R²   = {r2:.6f}"
)


print("\nGENERATED FILES")


print(
    "\n1.",
    H5_PATH
)


print(
    "\n2.",
    PKL_PATH
)


print(
    "\n3.",
    YAML_PATH
)


print(
    "\n4.",
    JSON_PATH
)


print("\n" + "=" * 80)
print("ALL FILES GENERATED SUCCESSFULLY")
print("=" * 80)
```
