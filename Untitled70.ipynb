```python
# =============================================================================
# AI-BASED FARMER INCOME PREDICTION
# MEMORY-EFFICIENT PARTICLE SWARM OPTIMIZATION (PSO)
#
# PSO is used for FEATURE SELECTION.
#
# MEMORY FIX:
#   - NO dense OneHotEncoder
#   - Numeric -> median imputation + StandardScaler
#   - Categorical -> most-frequent imputation + OrdinalEncoder
#   - float32 matrices
#   - ID / nearly unique categorical columns removed
#
# ALL OUTPUT FILES USE pso_ PREFIX
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import gc
import json
import yaml
import pickle
import zipfile
import tempfile
import shutil
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import (
    StandardScaler,
    OrdinalEncoder
)

from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    explained_variance_score
)

from sklearn.ensemble import ExtraTreesRegressor


# =============================================================================
# 2. GLOBAL SETTINGS
# =============================================================================

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =============================================================================
# 3. PROJECT DIRECTORY
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Farmer Income Prediction and Agricultural Socio-Economic Analysis"
)


DATA_SOURCE = os.path.join(
    PROJECT_DIR,
    "farmer-income-data.zip"
)


TRAIN_FILE = "lte_train.csv"
TEST_FILE = "lte_test.csv"
DICTIONARY_FILE = "lte_dictionary.csv"


TARGET_COLUMN = "Target_Variable/Total Income"


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. PSO OUTPUT PATHS
# =============================================================================

PSO_H5_PATH = os.path.join(
    PROJECT_DIR,
    "pso_model.h5"
)

PSO_PKL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_preprocessor.pkl"
)

PSO_YAML_PATH = os.path.join(
    PROJECT_DIR,
    "pso_config.yaml"
)

PSO_JSON_PATH = os.path.join(
    PROJECT_DIR,
    "pso_metadata.json"
)

PSO_SELECTED_FEATURES_PATH = os.path.join(
    PROJECT_DIR,
    "pso_selected_features.csv"
)

PSO_METRICS_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_model_metrics.csv"
)

PSO_RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result.csv"
)

PSO_PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction.csv"
)

PSO_ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_accuracy_graph.png"
)

PSO_COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_comparison_graph.png"
)

PSO_HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "pso_heatmap.png"
)

PSO_RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result_graph.png"
)

PSO_PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction_graph.png"
)

PSO_HISTORY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_training_history_graph.png"
)

PSO_CONVERGENCE_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_convergence_graph.png"
)

PSO_FEATURE_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection_graph.png"
)

PSO_RESIDUAL_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_residual_graph.png"
)

PSO_ACTUAL_PRED_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_actual_predicted_comparison_graph.png"
)


# =============================================================================
# 5. START
# =============================================================================

print("=" * 80)
print("AI-BASED FARMER INCOME PREDICTION")
print("MEMORY-EFFICIENT PARTICLE SWARM OPTIMIZATION")
print("=" * 80)

print("\nProject directory:")
print(PROJECT_DIR)


# =============================================================================
# 6. HELPER - FIND FILE IN DIRECTORY
# =============================================================================

def find_file_in_directory(directory, required_name):

    for root, dirs, files in os.walk(directory):

        for file_name in files:

            if file_name.lower() == required_name.lower():

                return os.path.join(
                    root,
                    file_name
                )

    return None


# =============================================================================
# 7. HELPER - FIND FILE IN ZIP
# =============================================================================

def find_file_in_zip(names, required_name):

    for name in names:

        if (
            os.path.basename(name).lower()
            ==
            required_name.lower()
        ):

            return name

    return None


# =============================================================================
# 8. LOAD DATASETS
# =============================================================================

def load_datasets():

    print("\nSearching for dataset files...")


    # =========================================================================
    # CASE 1: DATA SOURCE IS DIRECTORY
    # =========================================================================

    if os.path.isdir(DATA_SOURCE):

        train_path = find_file_in_directory(
            DATA_SOURCE,
            TRAIN_FILE
        )

        test_path = find_file_in_directory(
            DATA_SOURCE,
            TEST_FILE
        )

        dictionary_path = find_file_in_directory(
            DATA_SOURCE,
            DICTIONARY_FILE
        )

        if (
            train_path is not None
            and test_path is not None
            and dictionary_path is not None
        ):

            return (
                pd.read_csv(
                    train_path,
                    low_memory=False
                ),

                pd.read_csv(
                    test_path,
                    low_memory=False
                ),

                pd.read_csv(
                    dictionary_path,
                    low_memory=False
                )
            )


    # =========================================================================
    # CASE 2: ACTUAL ZIP FILE
    # =========================================================================

    if os.path.isfile(DATA_SOURCE):

        try:

            valid_zip = zipfile.is_zipfile(
                DATA_SOURCE
            )

        except Exception:

            valid_zip = False


        if valid_zip:

            try:

                print("Valid ZIP archive detected.")

                with zipfile.ZipFile(
                    DATA_SOURCE,
                    "r"
                ) as zip_ref:

                    names = zip_ref.namelist()

                    train_name = find_file_in_zip(
                        names,
                        TRAIN_FILE
                    )

                    test_name = find_file_in_zip(
                        names,
                        TEST_FILE
                    )

                    dictionary_name = find_file_in_zip(
                        names,
                        DICTIONARY_FILE
                    )

                    if train_name is None:
                        raise FileNotFoundError(TRAIN_FILE)

                    if test_name is None:
                        raise FileNotFoundError(TEST_FILE)

                    if dictionary_name is None:
                        raise FileNotFoundError(
                            DICTIONARY_FILE
                        )


                    with zip_ref.open(
                        train_name
                    ) as f:

                        train_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        test_name
                    ) as f:

                        test_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        dictionary_name
                    ) as f:

                        dictionary_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    return (
                        train_data,
                        test_data,
                        dictionary_data
                    )


            except PermissionError:

                print(
                    "\nPermission error while reading ZIP."
                )

                print(
                    "Creating temporary ZIP copy..."
                )


                temp_directory = tempfile.mkdtemp()

                temp_zip = os.path.join(
                    temp_directory,
                    "farmer-income-data.zip"
                )


                shutil.copy2(
                    DATA_SOURCE,
                    temp_zip
                )


                with zipfile.ZipFile(
                    temp_zip,
                    "r"
                ) as zip_ref:

                    names = zip_ref.namelist()

                    train_name = find_file_in_zip(
                        names,
                        TRAIN_FILE
                    )

                    test_name = find_file_in_zip(
                        names,
                        TEST_FILE
                    )

                    dictionary_name = find_file_in_zip(
                        names,
                        DICTIONARY_FILE
                    )


                    with zip_ref.open(
                        train_name
                    ) as f:

                        train_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        test_name
                    ) as f:

                        test_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                    with zip_ref.open(
                        dictionary_name
                    ) as f:

                        dictionary_data = pd.read_csv(
                            f,
                            low_memory=False
                        )


                shutil.rmtree(
                    temp_directory,
                    ignore_errors=True
                )


                return (
                    train_data,
                    test_data,
                    dictionary_data
                )


    # =========================================================================
    # CASE 3: SEARCH EXTRACTED CSV FILES
    # =========================================================================

    print(
        "Searching project directory for extracted CSV files..."
    )


    train_path = find_file_in_directory(
        PROJECT_DIR,
        TRAIN_FILE
    )

    test_path = find_file_in_directory(
        PROJECT_DIR,
        TEST_FILE
    )

    dictionary_path = find_file_in_directory(
        PROJECT_DIR,
        DICTIONARY_FILE
    )


    if train_path is None:
        raise FileNotFoundError(TRAIN_FILE)

    if test_path is None:
        raise FileNotFoundError(TEST_FILE)

    if dictionary_path is None:
        raise FileNotFoundError(DICTIONARY_FILE)


    return (
        pd.read_csv(
            train_path,
            low_memory=False
        ),

        pd.read_csv(
            test_path,
            low_memory=False
        ),

        pd.read_csv(
            dictionary_path,
            low_memory=False
        )
    )


# =============================================================================
# 9. LOAD DATA
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATA")
print("=" * 80)


train_df, test_df, dictionary_df = load_datasets()


print(
    "\nTraining shape:",
    train_df.shape
)

print(
    "Test shape:",
    test_df.shape
)

print(
    "Dictionary shape:",
    dictionary_df.shape
)


# =============================================================================
# 10. CLEAN COLUMN NAMES
# =============================================================================

train_df.columns = [
    str(column).strip()
    for column in train_df.columns
]

test_df.columns = [
    str(column).strip()
    for column in test_df.columns
]

dictionary_df.columns = [
    str(column).strip()
    for column in dictionary_df.columns
]


# =============================================================================
# 11. TARGET DETECTION
# =============================================================================

if TARGET_COLUMN not in train_df.columns:

    possible_targets = [

        column

        for column in train_df.columns

        if (
            "income" in str(column).lower()
            and
            "total" in str(column).lower()
        )
    ]


    if len(possible_targets) == 0:

        raise ValueError(
            "Could not find the total income target column."
        )


    TARGET_COLUMN = possible_targets[0]


print(
    "\nTarget column:",
    TARGET_COLUMN
)


# =============================================================================
# 12. CLEAN TARGET
# =============================================================================

train_df[
    TARGET_COLUMN
] = pd.to_numeric(
    train_df[
        TARGET_COLUMN
    ],
    errors="coerce"
)


train_df[
    TARGET_COLUMN
] = train_df[
    TARGET_COLUMN
].replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


train_df = train_df.dropna(
    subset=[
        TARGET_COLUMN
    ]
).reset_index(
    drop=True
)


# =============================================================================
# 13. CREATE X AND Y
# =============================================================================

X = train_df.drop(
    columns=[
        TARGET_COLUMN
    ]
).copy()


y = train_df[
    TARGET_COLUMN
].astype(
    np.float32
).copy()


if TARGET_COLUMN in test_df.columns:

    X_external_test = test_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    ).copy()

else:

    X_external_test = test_df.copy()


# =============================================================================
# 14. ALIGN TEST COLUMNS
# =============================================================================

for column in X.columns:

    if column not in X_external_test.columns:

        X_external_test[
            column
        ] = np.nan


X_external_test = X_external_test[
    X.columns
]


# =============================================================================
# 15. REMOVE ID / ALMOST-UNIQUE COLUMNS
# =============================================================================

print("\n" + "=" * 80)
print("CHECKING HIGH-CARDINALITY FEATURES")
print("=" * 80)


drop_columns = []


# Explicit identifier
if "FarmerID" in X.columns:

    drop_columns.append(
        "FarmerID"
    )


# Automatically identify nearly unique categorical fields
for column in X.columns:

    if X[column].dtype == "object":

        unique_count = X[
            column
        ].nunique(
            dropna=True
        )

        unique_ratio = (
            unique_count
            /
            max(
                len(X),
                1
            )
        )


        if (
            unique_count > 10000
            or
            unique_ratio > 0.80
        ):

            if column not in drop_columns:

                drop_columns.append(
                    column
                )


print(
    "\nColumns removed from modeling:"
)


if len(drop_columns) == 0:

    print("None")

else:

    for column in drop_columns:

        print(
            " -",
            column
        )


X = X.drop(
    columns=drop_columns,
    errors="ignore"
)


X_external_test = X_external_test.drop(
    columns=drop_columns,
    errors="ignore"
)


# =============================================================================
# 16. FEATURE TYPES
# =============================================================================

numeric_columns = X.select_dtypes(
    include=[
        np.number
    ]
).columns.tolist()


categorical_columns = [

    column

    for column in X.columns

    if column not in numeric_columns
]


print(
    "\nTotal usable features:",
    X.shape[1]
)

print(
    "Numeric features:",
    len(numeric_columns)
)

print(
    "Categorical features:",
    len(categorical_columns)
)


# =============================================================================
# 17. CLEAN NUMERIC FEATURES
# =============================================================================

for column in numeric_columns:

    X[
        column
    ] = pd.to_numeric(
        X[
            column
        ],
        errors="coerce"
    )


    X_external_test[
        column
    ] = pd.to_numeric(
        X_external_test[
            column
        ],
        errors="coerce"
    )


# =============================================================================
# 18. CLEAN CATEGORICAL FEATURES
# =============================================================================

for column in categorical_columns:

    X[
        column
    ] = X[
        column
    ].astype(
        "object"
    )


    X_external_test[
        column
    ] = X_external_test[
        column
    ].astype(
        "object"
    )


# =============================================================================
# 19. TRAIN / VALIDATION SPLIT
# =============================================================================

X_train, X_valid, y_train, y_valid = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42
)


print(
    "\nTraining records:",
    len(X_train)
)

print(
    "Validation records:",
    len(X_valid)
)


# =============================================================================
# 20. MEMORY-EFFICIENT PREPROCESSING
# =============================================================================

print("\n" + "=" * 80)
print("MEMORY-EFFICIENT PREPROCESSING")
print("=" * 80)


# =============================================================================
# 21. NUMERIC IMPUTATION
# =============================================================================

numeric_imputer = SimpleImputer(
    strategy="median"
)


if len(numeric_columns) > 0:

    X_train_numeric = numeric_imputer.fit_transform(
        X_train[
            numeric_columns
        ]
    )

    X_valid_numeric = numeric_imputer.transform(
        X_valid[
            numeric_columns
        ]
    )

    X_test_numeric = numeric_imputer.transform(
        X_external_test[
            numeric_columns
        ]
    )

else:

    X_train_numeric = np.empty(
        (
            len(X_train),
            0
        ),
        dtype=np.float32
    )

    X_valid_numeric = np.empty(
        (
            len(X_valid),
            0
        ),
        dtype=np.float32
    )

    X_test_numeric = np.empty(
        (
            len(X_external_test),
            0
        ),
        dtype=np.float32
    )


# =============================================================================
# 22. NUMERIC SCALING
# =============================================================================

numeric_scaler = StandardScaler()


if X_train_numeric.shape[1] > 0:

    X_train_numeric = numeric_scaler.fit_transform(
        X_train_numeric
    )

    X_valid_numeric = numeric_scaler.transform(
        X_valid_numeric
    )

    X_test_numeric = numeric_scaler.transform(
        X_test_numeric
    )


# =============================================================================
# 23. CATEGORICAL IMPUTATION
# =============================================================================

categorical_imputer = SimpleImputer(
    strategy="most_frequent"
)


if len(categorical_columns) > 0:

    X_train_categorical = categorical_imputer.fit_transform(
        X_train[
            categorical_columns
        ]
    )

    X_valid_categorical = categorical_imputer.transform(
        X_valid[
            categorical_columns
        ]
    )

    X_test_categorical = categorical_imputer.transform(
        X_external_test[
            categorical_columns
        ]
    )

else:

    X_train_categorical = np.empty(
        (
            len(X_train),
            0
        ),
        dtype=np.float32
    )

    X_valid_categorical = np.empty(
        (
            len(X_valid),
            0
        ),
        dtype=np.float32
    )

    X_test_categorical = np.empty(
        (
            len(X_external_test),
            0
        ),
        dtype=np.float32
    )


# =============================================================================
# 24. ORDINAL ENCODING
#
# IMPORTANT:
# Unlike OneHotEncoder, one categorical column remains one output column.
# This prevents the 40,000+ feature / 11.6 GB RAM problem.
# =============================================================================

categorical_encoder = OrdinalEncoder(

    handle_unknown="use_encoded_value",

    unknown_value=-1,

    encoded_missing_value=-1
)


if X_train_categorical.shape[1] > 0:

    X_train_categorical = categorical_encoder.fit_transform(
        X_train_categorical
    )

    X_valid_categorical = categorical_encoder.transform(
        X_valid_categorical
    )

    X_test_categorical = categorical_encoder.transform(
        X_test_categorical
    )


# =============================================================================
# 25. CONVERT TO FLOAT32
# =============================================================================

X_train_numeric = np.asarray(
    X_train_numeric,
    dtype=np.float32
)

X_valid_numeric = np.asarray(
    X_valid_numeric,
    dtype=np.float32
)

X_test_numeric = np.asarray(
    X_test_numeric,
    dtype=np.float32
)


X_train_categorical = np.asarray(
    X_train_categorical,
    dtype=np.float32
)

X_valid_categorical = np.asarray(
    X_valid_categorical,
    dtype=np.float32
)

X_test_categorical = np.asarray(
    X_test_categorical,
    dtype=np.float32
)


# =============================================================================
# 26. COMBINE FEATURES
# =============================================================================

X_train_processed = np.concatenate(
    [
        X_train_numeric,
        X_train_categorical
    ],
    axis=1
).astype(
    np.float32,
    copy=False
)


X_valid_processed = np.concatenate(
    [
        X_valid_numeric,
        X_valid_categorical
    ],
    axis=1
).astype(
    np.float32,
    copy=False
)


X_test_processed = np.concatenate(
    [
        X_test_numeric,
        X_test_categorical
    ],
    axis=1
).astype(
    np.float32,
    copy=False
)


processed_feature_names = (
    numeric_columns
    +
    categorical_columns
)


TOTAL_FEATURES = X_train_processed.shape[1]


print(
    "\nProcessed training shape:",
    X_train_processed.shape
)

print(
    "Processed validation shape:",
    X_valid_processed.shape
)

print(
    "Processed test shape:",
    X_test_processed.shape
)


# =============================================================================
# 27. MEMORY USAGE
# =============================================================================

print(
    f"\nTraining matrix memory: "
    f"{X_train_processed.nbytes / 1024 / 1024:.2f} MB"
)

print(
    f"Validation matrix memory: "
    f"{X_valid_processed.nbytes / 1024 / 1024:.2f} MB"
)

print(
    f"Test matrix memory: "
    f"{X_test_processed.nbytes / 1024 / 1024:.2f} MB"
)


# =============================================================================
# 28. RELEASE INTERMEDIATE ARRAYS
# =============================================================================

del X_train_numeric
del X_valid_numeric
del X_test_numeric

del X_train_categorical
del X_valid_categorical
del X_test_categorical

gc.collect()


# =============================================================================
# 29. TARGET SCALER
# =============================================================================

target_scaler = StandardScaler()


y_train_array = np.asarray(
    y_train,
    dtype=np.float32
)

y_valid_array = np.asarray(
    y_valid,
    dtype=np.float32
)


y_train_scaled = target_scaler.fit_transform(
    y_train_array.reshape(
        -1,
        1
    )
).flatten().astype(
    np.float32
)


y_valid_scaled = target_scaler.transform(
    y_valid_array.reshape(
        -1,
        1
    )
).flatten().astype(
    np.float32
)


# =============================================================================
# 30. PSO SETTINGS
# =============================================================================

print("\n" + "=" * 80)
print("PARTICLE SWARM OPTIMIZATION CONFIGURATION")
print("=" * 80)


PSO_CANDIDATE_POOL_SIZE = min(
    60,
    TOTAL_FEATURES
)


PSO_MIN_FEATURES = min(
    8,
    PSO_CANDIDATE_POOL_SIZE
)


PSO_MAX_FEATURES = min(
    35,
    PSO_CANDIDATE_POOL_SIZE
)


PSO_PARTICLES = 20

PSO_ITERATIONS = 20


# Inertia weight
W_MAX = 0.90
W_MIN = 0.40


# Cognitive coefficient
C1 = 2.0


# Social coefficient
C2 = 2.0


# Maximum velocity
V_MAX = 6.0


print(
    "\nTotal processed features:",
    TOTAL_FEATURES
)

print(
    "PSO candidate pool:",
    PSO_CANDIDATE_POOL_SIZE
)

print(
    "Particles:",
    PSO_PARTICLES
)

print(
    "Iterations:",
    PSO_ITERATIONS
)

print(
    "Minimum selected features:",
    PSO_MIN_FEATURES
)

print(
    "Maximum selected features:",
    PSO_MAX_FEATURES
)


# =============================================================================
# 31. FEATURE PRE-SCREENING
#
# ExtraTrees is used only to reduce the feature search space.
# PSO determines the final feature subset.
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PSO CANDIDATE FEATURE POOL")
print("=" * 80)


screening_model = ExtraTreesRegressor(

    n_estimators=120,

    random_state=42,

    n_jobs=-1,

    max_depth=12,

    max_features="sqrt"
)


screening_model.fit(
    X_train_processed,
    y_train_array
)


feature_importances = screening_model.feature_importances_


candidate_indices = np.argsort(
    feature_importances
)[::-1][
    :PSO_CANDIDATE_POOL_SIZE
]


candidate_feature_names = [

    processed_feature_names[index]

    for index in candidate_indices
]


print(
    "\nCandidate features:"
)


for index, feature in enumerate(
    candidate_feature_names,
    start=1
):

    print(
        f"{index:02d}. {feature}"
    )


# =============================================================================
# 32. CREATE PSO CANDIDATE MATRICES
# =============================================================================

X_train_candidate = X_train_processed[
    :,
    candidate_indices
].astype(
    np.float32,
    copy=False
)


X_valid_candidate = X_valid_processed[
    :,
    candidate_indices
].astype(
    np.float32,
    copy=False
)


X_test_candidate = X_test_processed[
    :,
    candidate_indices
].astype(
    np.float32,
    copy=False
)


# =============================================================================
# 33. RANDOM GENERATOR
# =============================================================================

rng = np.random.default_rng(
    42
)


# =============================================================================
# 34. REPAIR PARTICLE
#
# Makes sure each particle selects between the configured minimum and
# maximum number of features.
# =============================================================================

def repair_particle(position):

    position = np.asarray(
        position,
        dtype=np.int8
    ).copy()


    selected = np.where(
        position == 1
    )[0]


    # -------------------------------------------------------------------------
    # Too few features
    # -------------------------------------------------------------------------

    if len(selected) < PSO_MIN_FEATURES:

        zero_indices = np.where(
            position == 0
        )[0]


        required = (
            PSO_MIN_FEATURES
            -
            len(selected)
        )


        if len(zero_indices) > 0:

            additions = rng.choice(

                zero_indices,

                size=min(
                    required,
                    len(zero_indices)
                ),

                replace=False
            )


            position[
                additions
            ] = 1


    # -------------------------------------------------------------------------
    # Too many features
    # -------------------------------------------------------------------------

    selected = np.where(
        position == 1
    )[0]


    if len(selected) > PSO_MAX_FEATURES:

        remove_count = (
            len(selected)
            -
            PSO_MAX_FEATURES
        )


        removals = rng.choice(

            selected,

            size=remove_count,

            replace=False
        )


        position[
            removals
        ] = 0


    return position


# =============================================================================
# 35. PSO FITNESS CACHE
# =============================================================================

fitness_cache = {}


# =============================================================================
# 36. FITNESS FUNCTION
#
# PSO searches for features that maximize validation R².
#
# A small penalty is included for unnecessarily large feature subsets.
# =============================================================================

def evaluate_particle(position):

    position = repair_particle(
        position
    )


    key = tuple(
        position.tolist()
    )


    if key in fitness_cache:

        return fitness_cache[key]


    selected = np.where(
        position == 1
    )[0]


    if len(selected) < PSO_MIN_FEATURES:

        return -999.0


    Xtr = X_train_candidate[
        :,
        selected
    ]


    Xva = X_valid_candidate[
        :,
        selected
    ]


    surrogate_model = ExtraTreesRegressor(

        n_estimators=60,

        random_state=42,

        n_jobs=-1,

        max_depth=10,

        max_features="sqrt"
    )


    surrogate_model.fit(
        Xtr,
        y_train_array
    )


    predictions = surrogate_model.predict(
        Xva
    )


    validation_r2 = r2_score(
        y_valid_array,
        predictions
    )


    complexity_penalty = (

        len(selected)

        /

        PSO_CANDIDATE_POOL_SIZE

        *

        0.02
    )


    fitness = (
        validation_r2
        -
        complexity_penalty
    )


    fitness_cache[
        key
    ] = float(
        fitness
    )


    return float(
        fitness
    )


# =============================================================================
# 37. INITIALIZE BINARY PARTICLES
# =============================================================================

print("\n" + "=" * 80)
print("INITIALIZING PSO PARTICLES")
print("=" * 80)


positions = np.zeros(
    (
        PSO_PARTICLES,
        PSO_CANDIDATE_POOL_SIZE
    ),
    dtype=np.int8
)


velocities = rng.uniform(

    low=-1.0,

    high=1.0,

    size=(
        PSO_PARTICLES,
        PSO_CANDIDATE_POOL_SIZE
    )
).astype(
    np.float32
)


for particle_index in range(
    PSO_PARTICLES
):

    number_features = int(
        rng.integers(
            PSO_MIN_FEATURES,
            PSO_MAX_FEATURES + 1
        )
    )


    selected_indices = rng.choice(

        PSO_CANDIDATE_POOL_SIZE,

        size=number_features,

        replace=False
    )


    positions[
        particle_index,
        selected_indices
    ] = 1


# =============================================================================
# 38. INITIAL FITNESS
# =============================================================================

personal_best_positions = positions.copy()


personal_best_scores = np.full(
    PSO_PARTICLES,
    -np.inf,
    dtype=np.float64
)


global_best_position = None

global_best_score = -np.inf


for particle_index in range(
    PSO_PARTICLES
):

    score = evaluate_particle(
        positions[
            particle_index
        ]
    )


    personal_best_scores[
        particle_index
    ] = score


    personal_best_positions[
        particle_index
    ] = positions[
        particle_index
    ].copy()


    if score > global_best_score:

        global_best_score = score

        global_best_position = positions[
            particle_index
        ].copy()


print(
    "\nInitial global best fitness:",
    global_best_score
)


# =============================================================================
# 39. PSO HISTORY
# =============================================================================

pso_fitness_history = []

pso_feature_count_history = []

pso_mean_fitness_history = []


# =============================================================================
# 40. SIGMOID FUNCTION FOR BINARY PSO
# =============================================================================

def sigmoid(x):

    x = np.clip(
        x,
        -20,
        20
    )

    return (
        1.0
        /
        (
            1.0
            +
            np.exp(
                -x
            )
        )
    )


# =============================================================================
# 41. RUN BINARY PSO
# =============================================================================

print("\n" + "=" * 80)
print("RUNNING BINARY PARTICLE SWARM OPTIMIZATION")
print("=" * 80)


for iteration in range(
    PSO_ITERATIONS
):

    # -------------------------------------------------------------------------
    # Decrease inertia from W_MAX to W_MIN
    # -------------------------------------------------------------------------

    if PSO_ITERATIONS > 1:

        inertia = (
            W_MAX
            -
            (
                W_MAX
                -
                W_MIN
            )
            *
            (
                iteration
                /
                (
                    PSO_ITERATIONS
                    -
                    1
                )
            )
        )

    else:

        inertia = W_MIN


    iteration_scores = []


    # =========================================================================
    # UPDATE EVERY PARTICLE
    # =========================================================================

    for particle_index in range(
        PSO_PARTICLES
    ):

        r1 = rng.random(
            PSO_CANDIDATE_POOL_SIZE
        )

        r2 = rng.random(
            PSO_CANDIDATE_POOL_SIZE
        )


        cognitive_component = (

            C1

            *

            r1

            *

            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )
        )


        social_component = (

            C2

            *

            r2

            *

            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )
        )


        velocities[
            particle_index
        ] = (

            inertia

            *

            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )


        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -V_MAX,

            V_MAX
        )


        probabilities = sigmoid(
            velocities[
                particle_index
            ]
        )


        random_values = rng.random(
            PSO_CANDIDATE_POOL_SIZE
        )


        new_position = (
            random_values
            <
            probabilities
        ).astype(
            np.int8
        )


        new_position = repair_particle(
            new_position
        )


        positions[
            particle_index
        ] = new_position


        # ---------------------------------------------------------------------
        # Evaluate new position
        # ---------------------------------------------------------------------

        score = evaluate_particle(
            new_position
        )


        iteration_scores.append(
            score
        )


        # ---------------------------------------------------------------------
        # Personal best
        # ---------------------------------------------------------------------

        if (
            score
            >
            personal_best_scores[
                particle_index
            ]
        ):

            personal_best_scores[
                particle_index
            ] = score


            personal_best_positions[
                particle_index
            ] = new_position.copy()


        # ---------------------------------------------------------------------
        # Global best
        # ---------------------------------------------------------------------

        if score > global_best_score:

            global_best_score = score

            global_best_position = new_position.copy()


    # =========================================================================
    # SAVE ITERATION HISTORY
    # =========================================================================

    pso_fitness_history.append(
        float(
            global_best_score
        )
    )


    pso_feature_count_history.append(
        int(
            global_best_position.sum()
        )
    )


    pso_mean_fitness_history.append(
        float(
            np.mean(
                iteration_scores
            )
        )
    )


    print(

        f"Iteration "
        f"{iteration + 1:02d}/"
        f"{PSO_ITERATIONS} | "

        f"Global Best = "
        f"{global_best_score:.6f} | "

        f"Mean Fitness = "
        f"{np.mean(iteration_scores):.6f} | "

        f"Features = "
        f"{int(global_best_position.sum())} | "

        f"Inertia = "
        f"{inertia:.3f}"
    )


    gc.collect()


# =============================================================================
# 42. PSO COMPLETE
# =============================================================================

print("\n" + "=" * 80)
print("PSO OPTIMIZATION COMPLETED")
print("=" * 80)


selected_candidate_positions = np.where(
    global_best_position == 1
)[0]


selected_original_indices = candidate_indices[
    selected_candidate_positions
]


selected_feature_names = [

    processed_feature_names[
        index
    ]

    for index in selected_original_indices
]


selected_importances = feature_importances[
    selected_original_indices
]


print(
    "\nBest PSO fitness:",
    global_best_score
)


print(
    "Selected features:",
    len(
        selected_feature_names
    )
)


print(
    "\nPSO-selected feature list:"
)


for index, feature in enumerate(
    selected_feature_names,
    start=1
):

    print(
        f"{index:02d}. {feature}"
    )


# =============================================================================
# 43. SAVE SELECTED FEATURES
# =============================================================================

selected_features_df = pd.DataFrame(
    {
        "Feature_Name":
            selected_feature_names,

        "Original_Feature_Index":
            selected_original_indices,

        "Screening_Importance":
            selected_importances
    }
)


selected_features_df = (
    selected_features_df
    .sort_values(
        by="Screening_Importance",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


selected_features_df.to_csv(
    PSO_SELECTED_FEATURES_PATH,
    index=False
)


# =============================================================================
# 44. FINAL PSO MATRICES
# =============================================================================

X_train_pso = X_train_processed[
    :,
    selected_original_indices
].astype(
    np.float32,
    copy=False
)


X_valid_pso = X_valid_processed[
    :,
    selected_original_indices
].astype(
    np.float32,
    copy=False
)


X_test_pso = X_test_processed[
    :,
    selected_original_indices
].astype(
    np.float32,
    copy=False
)


print(
    "\nFinal PSO training shape:",
    X_train_pso.shape
)

print(
    "Final PSO validation shape:",
    X_valid_pso.shape
)

print(
    "Final PSO test shape:",
    X_test_pso.shape
)


# =============================================================================
# 45. RELEASE PSO SEARCH MATRICES
# =============================================================================

del X_train_candidate
del X_valid_candidate
del X_test_candidate

del screening_model

gc.collect()


# =============================================================================
# 46. BUILD PSO-OPTIMIZED ANN
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING PSO-OPTIMIZED NEURAL NETWORK")
print("=" * 80)


INPUT_DIMENSION = X_train_pso.shape[1]


pso_model = Sequential(
    [

        Input(
            shape=(
                INPUT_DIMENSION,
            )
        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.25
        ),

        Dense(
            64,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            0.20
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            0.10
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )
    ]
)


pso_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=[
        "mae"
    ]
)


pso_model.summary()


# =============================================================================
# 47. CALLBACKS
# =============================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=15,

    restore_best_weights=True,

    verbose=1
)


reduce_lr = ReduceLROnPlateau(

    monitor="val_loss",

    factor=0.5,

    patience=6,

    min_lr=1e-6,

    verbose=1
)


# =============================================================================
# 48. TRAIN PSO MODEL
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING PSO-OPTIMIZED MODEL")
print("=" * 80)


history = pso_model.fit(

    X_train_pso,

    y_train_scaled,

    validation_data=(
        X_valid_pso,
        y_valid_scaled
    ),

    epochs=150,

    batch_size=128,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)


# =============================================================================
# 49. TRAINING PREDICTIONS
# =============================================================================

print("\n" + "=" * 80)
print("GENERATING PSO PREDICTIONS")
print("=" * 80)


train_prediction_scaled = pso_model.predict(
    X_train_pso,
    batch_size=512,
    verbose=0
)


train_predictions = (
    target_scaler
    .inverse_transform(
        train_prediction_scaled
    )
    .flatten()
)


# =============================================================================
# 50. VALIDATION PREDICTIONS
# =============================================================================

valid_prediction_scaled = pso_model.predict(
    X_valid_pso,
    batch_size=512,
    verbose=0
)


valid_predictions = (
    target_scaler
    .inverse_transform(
        valid_prediction_scaled
    )
    .flatten()
)


# =============================================================================
# 51. TEST PREDICTIONS
# =============================================================================

test_prediction_scaled = pso_model.predict(
    X_test_pso,
    batch_size=512,
    verbose=0
)


test_predictions = (
    target_scaler
    .inverse_transform(
        test_prediction_scaled
    )
    .flatten()
)


# =============================================================================
# 52. TRAINING METRICS
# =============================================================================

train_mae = mean_absolute_error(
    y_train_array,
    train_predictions
)


train_mse = mean_squared_error(
    y_train_array,
    train_predictions
)


train_rmse = np.sqrt(
    train_mse
)


train_r2 = r2_score(
    y_train_array,
    train_predictions
)


train_ev = explained_variance_score(
    y_train_array,
    train_predictions
)


# =============================================================================
# 53. VALIDATION METRICS
# =============================================================================

valid_mae = mean_absolute_error(
    y_valid_array,
    valid_predictions
)


valid_mse = mean_squared_error(
    y_valid_array,
    valid_predictions
)


valid_rmse = np.sqrt(
    valid_mse
)


valid_r2 = r2_score(
    y_valid_array,
    valid_predictions
)


valid_ev = explained_variance_score(
    y_valid_array,
    valid_predictions
)


# =============================================================================
# 54. REGRESSION SCORE
#
# This is NOT classification accuracy.
#
# It is a normalized RMSE-based regression score for visualization.
# R² remains the primary regression goodness-of-fit metric.
# =============================================================================

train_range = (
    np.max(
        y_train_array
    )
    -
    np.min(
        y_train_array
    )
)


valid_range = (
    np.max(
        y_valid_array
    )
    -
    np.min(
        y_valid_array
    )
)


if train_range > 0:

    train_score = (
        1
        -
        train_rmse
        /
        train_range
    )

else:

    train_score = 0.0


if valid_range > 0:

    valid_score = (
        1
        -
        valid_rmse
        /
        valid_range
    )

else:

    valid_score = 0.0


train_score = np.clip(
    train_score,
    0,
    1
)


valid_score = np.clip(
    valid_score,
    0,
    1
)


# =============================================================================
# 55. PRINT PERFORMANCE
# =============================================================================

print("\n" + "=" * 80)
print("PSO MODEL PERFORMANCE")
print("=" * 80)


print("\nTRAINING")


print(
    f"MAE                 : "
    f"{train_mae:,.4f}"
)

print(
    f"MSE                 : "
    f"{train_mse:,.4f}"
)

print(
    f"RMSE                : "
    f"{train_rmse:,.4f}"
)

print(
    f"R²                  : "
    f"{train_r2:.6f}"
)

print(
    f"Explained Variance  : "
    f"{train_ev:.6f}"
)

print(
    f"Regression Score    : "
    f"{train_score * 100:.2f}%"
)


print("\nVALIDATION")


print(
    f"MAE                 : "
    f"{valid_mae:,.4f}"
)

print(
    f"MSE                 : "
    f"{valid_mse:,.4f}"
)

print(
    f"RMSE                : "
    f"{valid_rmse:,.4f}"
)

print(
    f"R²                  : "
    f"{valid_r2:.6f}"
)

print(
    f"Explained Variance  : "
    f"{valid_ev:.6f}"
)

print(
    f"Regression Score    : "
    f"{valid_score * 100:.2f}%"
)


# =============================================================================
# 56. SAVE PSO MODEL
# =============================================================================

print("\n" + "=" * 80)
print("SAVING PSO MODEL")
print("=" * 80)


pso_model.save(
    PSO_H5_PATH
)


print(
    "\nSaved:",
    PSO_H5_PATH
)


# =============================================================================
# 57. SAVE PSO PREPROCESSOR PKL
# =============================================================================

pso_package = {

    "numeric_columns":
        numeric_columns,

    "categorical_columns":
        categorical_columns,

    "dropped_columns":
        drop_columns,

    "numeric_imputer":
        numeric_imputer,

    "numeric_scaler":
        numeric_scaler,

    "categorical_imputer":
        categorical_imputer,

    "categorical_encoder":
        categorical_encoder,

    "target_scaler":
        target_scaler,

    "target_column":
        TARGET_COLUMN,

    "processed_feature_names":
        processed_feature_names,

    "candidate_indices":
        candidate_indices,

    "candidate_feature_names":
        candidate_feature_names,

    "selected_feature_indices":
        selected_original_indices,

    "selected_feature_names":
        selected_feature_names,

    "global_best_position":
        global_best_position,

    "global_best_fitness":
        float(
            global_best_score
        ),

    "pso_fitness_history":
        pso_fitness_history,

    "pso_feature_count_history":
        pso_feature_count_history
}


with open(
    PSO_PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        pso_package,
        file
    )


print(
    "Saved:",
    PSO_PKL_PATH
)


# =============================================================================
# 58. SAVE YAML
# =============================================================================

yaml_data = {

    "project": {

        "name":
            "AI-Based Farmer Income Prediction and "
            "Agricultural Socio-Economic Analysis",

        "method":
            "Binary Particle Swarm Optimization",

        "target":
            str(
                TARGET_COLUMN
            ),

        "problem_type":
            "Regression"
    },

    "preprocessing": {

        "categorical_encoding":
            "OrdinalEncoder",

        "numeric_scaling":
            "StandardScaler",

        "numeric_imputation":
            "Median",

        "categorical_imputation":
            "Most Frequent",

        "dtype":
            "float32",

        "total_processed_features":
            int(
                TOTAL_FEATURES
            )
    },

    "pso": {

        "particles":
            int(
                PSO_PARTICLES
            ),

        "iterations":
            int(
                PSO_ITERATIONS
            ),

        "candidate_pool_size":
            int(
                PSO_CANDIDATE_POOL_SIZE
            ),

        "minimum_features":
            int(
                PSO_MIN_FEATURES
            ),

        "maximum_features":
            int(
                PSO_MAX_FEATURES
            ),

        "cognitive_coefficient":
            float(
                C1
            ),

        "social_coefficient":
            float(
                C2
            ),

        "inertia_max":
            float(
                W_MAX
            ),

        "inertia_min":
            float(
                W_MIN
            ),

        "selected_features":
            int(
                len(
                    selected_feature_names
                )
            ),

        "best_fitness":
            float(
                global_best_score
            )
    },

    "validation_metrics": {

        "MAE":
            float(
                valid_mae
            ),

        "MSE":
            float(
                valid_mse
            ),

        "RMSE":
            float(
                valid_rmse
            ),

        "R2":
            float(
                valid_r2
            ),

        "Explained_Variance":
            float(
                valid_ev
            )
    }
}


with open(
    PSO_YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False
    )


print(
    "Saved:",
    PSO_YAML_PATH
)


# =============================================================================
# 59. SAVE JSON
# =============================================================================

json_data = {

    "project":
        "AI-Based Farmer Income Prediction",

    "optimization":
        "Binary Particle Swarm Optimization",

    "memory_efficient":
        True,

    "encoding":
        "OrdinalEncoder",

    "processed_features":
        int(
            TOTAL_FEATURES
        ),

    "candidate_features":
        int(
            PSO_CANDIDATE_POOL_SIZE
        ),

    "selected_features":
        int(
            len(
                selected_feature_names
            )
        ),

    "selected_feature_names": [

        str(feature)

        for feature in selected_feature_names
    ],

    "best_pso_fitness":
        float(
            global_best_score
        ),

    "training_metrics": {

        "MAE":
            float(
                train_mae
            ),

        "MSE":
            float(
                train_mse
            ),

        "RMSE":
            float(
                train_rmse
            ),

        "R2":
            float(
                train_r2
            ),

        "Explained_Variance":
            float(
                train_ev
            )
    },

    "validation_metrics": {

        "MAE":
            float(
                valid_mae
            ),

        "MSE":
            float(
                valid_mse
            ),

        "RMSE":
            float(
                valid_rmse
            ),

        "R2":
            float(
                valid_r2
            ),

        "Explained_Variance":
            float(
                valid_ev
            )
    }
}


with open(
    PSO_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4
    )


print(
    "Saved:",
    PSO_JSON_PATH
)


# =============================================================================
# 60. SAVE METRICS CSV
# =============================================================================

metrics_df = pd.DataFrame(
    {

        "Dataset": [
            "Training",
            "Validation"
        ],

        "MAE": [
            train_mae,
            valid_mae
        ],

        "MSE": [
            train_mse,
            valid_mse
        ],

        "RMSE": [
            train_rmse,
            valid_rmse
        ],

        "R2": [
            train_r2,
            valid_r2
        ],

        "Explained_Variance": [
            train_ev,
            valid_ev
        ],

        "Regression_Score_Percent": [
            train_score * 100,
            valid_score * 100
        ]
    }
)


metrics_df.to_csv(
    PSO_METRICS_CSV_PATH,
    index=False
)


# =============================================================================
# 61. PSO ACCURACY / PERFORMANCE GRAPH
# =============================================================================

print(
    "\nCreating pso_accuracy_graph.png..."
)


metric_names = [
    "R²",
    "Explained\nVariance",
    "Regression\nScore"
]


training_scores = [

    np.clip(
        train_r2,
        0,
        1
    ) * 100,

    np.clip(
        train_ev,
        0,
        1
    ) * 100,

    train_score * 100
]


validation_scores = [

    np.clip(
        valid_r2,
        0,
        1
    ) * 100,

    np.clip(
        valid_ev,
        0,
        1
    ) * 100,

    valid_score * 100
]


x_axis = np.arange(
    len(
        metric_names
    )
)


width = 0.35


plt.figure(
    figsize=(
        10,
        6
    )
)


bars1 = plt.bar(
    x_axis - width / 2,
    training_scores,
    width,
    label="Training"
)


bars2 = plt.bar(
    x_axis + width / 2,
    validation_scores,
    width,
    label="Validation"
)


plt.xlabel(
    "Regression Performance Metric"
)

plt.ylabel(
    "Score (%)"
)

plt.title(
    "PSO Farmer Income Prediction Performance"
)

plt.xticks(
    x_axis,
    metric_names
)

plt.ylim(
    0,
    105
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.25
)


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        value = bar.get_height()

        plt.text(
            bar.get_x()
            +
            bar.get_width() / 2,

            value + 1,

            f"{value:.1f}%",

            ha="center"
        )


plt.tight_layout()


plt.savefig(
    PSO_ACCURACY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 62. PSO COMPARISON GRAPH
# =============================================================================

print(
    "Creating pso_comparison_graph.png..."
)


metrics = [
    "MAE",
    "RMSE"
]


training_errors = [
    train_mae,
    train_rmse
]


validation_errors = [
    valid_mae,
    valid_rmse
]


x_axis = np.arange(
    len(metrics)
)


width = 0.35


plt.figure(
    figsize=(
        9,
        6
    )
)


plt.bar(
    x_axis - width / 2,
    training_errors,
    width,
    label="Training"
)


plt.bar(
    x_axis + width / 2,
    validation_errors,
    width,
    label="Validation"
)


plt.xlabel(
    "Metric"
)

plt.ylabel(
    "Error"
)

plt.title(
    "PSO Training vs Validation Error"
)

plt.xticks(
    x_axis,
    metrics
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PSO_COMPARISON_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 63. CORRELATION HEATMAP
# =============================================================================

print(
    "Creating pso_heatmap.png..."
)


numeric_heatmap_df = train_df.select_dtypes(
    include=[
        np.number
    ]
).copy()


numeric_heatmap_df = numeric_heatmap_df.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


numeric_heatmap_df = numeric_heatmap_df.dropna(
    axis=1,
    how="all"
)


if TARGET_COLUMN in numeric_heatmap_df.columns:

    target_correlations = (
        numeric_heatmap_df
        .corr()[
            TARGET_COLUMN
        ]
        .abs()
        .sort_values(
            ascending=False
        )
    )


    heatmap_columns = (
        target_correlations
        .head(12)
        .index
        .tolist()
    )

else:

    heatmap_columns = (
        numeric_heatmap_df
        .columns[:12]
        .tolist()
    )


correlation_matrix = numeric_heatmap_df[
    heatmap_columns
].corr()


fig, ax = plt.subplots(
    figsize=(
        14,
        11
    )
)


image = ax.imshow(
    correlation_matrix.values,
    vmin=-1,
    vmax=1,
    aspect="auto"
)


colorbar = fig.colorbar(
    image,
    ax=ax
)


colorbar.set_label(
    "Correlation"
)


ax.set_xticks(
    np.arange(
        len(
            heatmap_columns
        )
    )
)


ax.set_yticks(
    np.arange(
        len(
            heatmap_columns
        )
    )
)


ax.set_xticklabels(
    heatmap_columns,
    rotation=90,
    fontsize=8
)


ax.set_yticklabels(
    heatmap_columns,
    fontsize=8
)


for row in range(
    len(
        heatmap_columns
    )
):

    for column in range(
        len(
            heatmap_columns
        )
    ):

        value = correlation_matrix.iloc[
            row,
            column
        ]


        if not np.isnan(value):

            ax.text(
                column,
                row,
                f"{value:.2f}",
                ha="center",
                va="center",
                fontsize=7
            )


ax.set_title(
    "PSO Farmer Income Correlation Heatmap"
)


plt.tight_layout()


plt.savefig(
    PSO_HEATMAP_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 64. RESULT CSV
# =============================================================================

result_df = pd.DataFrame(
    {

        "Actual_Income":
            y_valid_array,

        "PSO_Predicted_Income":
            valid_predictions
    }
)


result_df[
    "Residual"
] = (
    result_df[
        "Actual_Income"
    ]
    -
    result_df[
        "PSO_Predicted_Income"
    ]
)


result_df[
    "Absolute_Error"
] = np.abs(
    result_df[
        "Residual"
    ]
)


actual_absolute = np.abs(
    result_df[
        "Actual_Income"
    ].values
)


result_df[
    "Percentage_Error"
] = np.where(

    actual_absolute > 1e-8,

    (
        result_df[
            "Absolute_Error"
        ].values
        /
        actual_absolute
    )
    * 100,

    np.nan
)


result_df.to_csv(
    PSO_RESULT_CSV_PATH,
    index=False
)


print(
    "\nFirst 10 PSO validation results:"
)


print(
    result_df.head(
        10
    ).to_string(
        index=False
    )
)


# =============================================================================
# 65. RESULT GRAPH
# =============================================================================

print(
    "Creating pso_result_graph.png..."
)


plt.figure(
    figsize=(
        10,
        7
    )
)


plt.scatter(
    result_df[
        "Actual_Income"
    ],

    result_df[
        "PSO_Predicted_Income"
    ],

    alpha=0.45,
    s=20
)


minimum_value = min(
    result_df[
        "Actual_Income"
    ].min(),

    result_df[
        "PSO_Predicted_Income"
    ].min()
)


maximum_value = max(
    result_df[
        "Actual_Income"
    ].max(),

    result_df[
        "PSO_Predicted_Income"
    ].max()
)


plt.plot(
    [
        minimum_value,
        maximum_value
    ],

    [
        minimum_value,
        maximum_value
    ],

    linestyle="--",

    linewidth=2,

    label="Perfect Prediction"
)


plt.xlabel(
    "Actual Farmer Income"
)

plt.ylabel(
    "PSO Predicted Farmer Income"
)

plt.title(
    "PSO Actual vs Predicted Farmer Income"
)

plt.legend()

plt.grid(
    alpha=0.25
)


performance_text = (
    f"R² = {valid_r2:.4f}\n"
    f"MAE = {valid_mae:,.2f}\n"
    f"RMSE = {valid_rmse:,.2f}"
)


plt.text(
    0.05,
    0.95,
    performance_text,
    transform=plt.gca().transAxes,
    verticalalignment="top",
    bbox=dict(
        boxstyle="round",
        alpha=0.2
    )
)


plt.tight_layout()


plt.savefig(
    PSO_RESULT_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 66. PREDICTION CSV
# =============================================================================

prediction_df = test_df.copy()


if TARGET_COLUMN in prediction_df.columns:

    prediction_df = prediction_df.drop(
        columns=[
            TARGET_COLUMN
        ]
    )


prediction_df[
    "PSO_Predicted_Total_Income"
] = test_predictions


prediction_df.insert(
    0,
    "Prediction_ID",
    np.arange(
        1,
        len(prediction_df) + 1
    )
)


prediction_df.to_csv(
    PSO_PREDICTION_CSV_PATH,
    index=False
)


print(
    "\nFirst 10 PSO test predictions:"
)


print(
    prediction_df[
        [
            "Prediction_ID",
            "PSO_Predicted_Total_Income"
        ]
    ]
    .head(
        10
    )
    .to_string(
        index=False
    )
)


# =============================================================================
# 67. PREDICTION GRAPH
# =============================================================================

print(
    "Creating pso_prediction_graph.png..."
)


sorted_predictions = np.sort(
    test_predictions
)


prediction_numbers = np.arange(
    1,
    len(sorted_predictions) + 1
)


mean_prediction = np.mean(
    test_predictions
)


plt.figure(
    figsize=(
        12,
        6
    )
)


plt.plot(
    prediction_numbers,
    sorted_predictions,
    linewidth=2,
    label="PSO Predicted Income"
)


plt.axhline(
    mean_prediction,
    linestyle="--",
    linewidth=2,
    label=(
        f"Mean = "
        f"{mean_prediction:,.2f}"
    )
)


plt.xlabel(
    "Test Records Sorted by Predicted Income"
)

plt.ylabel(
    "Predicted Farmer Income"
)

plt.title(
    "PSO Farmer Income Prediction Distribution"
)

plt.legend()

plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PSO_PREDICTION_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 68. TRAINING HISTORY GRAPH
# =============================================================================

print(
    "Creating pso_training_history_graph.png..."
)


training_loss = history.history[
    "loss"
]


validation_loss = history.history[
    "val_loss"
]


epochs = np.arange(
    1,
    len(training_loss) + 1
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.plot(
    epochs,
    training_loss,
    linewidth=2,
    label="Training Loss"
)


plt.plot(
    epochs,
    validation_loss,
    linewidth=2,
    label="Validation Loss"
)


plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "MSE Loss"
)

plt.title(
    "PSO-Optimized ANN Training History"
)

plt.legend()

plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PSO_HISTORY_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 69. PSO CONVERGENCE GRAPH
# =============================================================================

print(
    "Creating pso_convergence_graph.png..."
)


iteration_numbers = np.arange(
    1,
    PSO_ITERATIONS + 1
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.plot(
    iteration_numbers,
    pso_fitness_history,
    marker="o",
    linewidth=2,
    label="Global Best Fitness"
)


plt.plot(
    iteration_numbers,
    pso_mean_fitness_history,
    marker="s",
    linewidth=2,
    label="Mean Swarm Fitness"
)


plt.xlabel(
    "PSO Iteration"
)

plt.ylabel(
    "Fitness"
)

plt.title(
    "Particle Swarm Optimization Convergence"
)

plt.legend()

plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PSO_CONVERGENCE_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 70. PSO FEATURE SELECTION GRAPH
# =============================================================================

print(
    "Creating pso_feature_selection_graph.png..."
)


feature_plot_df = (
    selected_features_df
    .sort_values(
        by="Screening_Importance",
        ascending=True
    )
)


plt.figure(
    figsize=(
        12,
        max(
            7,
            len(feature_plot_df) * 0.35
        )
    )
)


plt.barh(
    feature_plot_df[
        "Feature_Name"
    ],

    feature_plot_df[
        "Screening_Importance"
    ]
)


plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "PSO Selected Feature"
)

plt.title(
    "PSO Selected Features for Farmer Income Prediction"
)


plt.tight_layout()


plt.savefig(
    PSO_FEATURE_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 71. RESIDUAL GRAPH
# =============================================================================

print(
    "Creating pso_residual_graph.png..."
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.scatter(
    result_df[
        "PSO_Predicted_Income"
    ],

    result_df[
        "Residual"
    ],

    alpha=0.45,
    s=20
)


plt.axhline(
    0,
    linestyle="--",
    linewidth=2
)


plt.xlabel(
    "PSO Predicted Farmer Income"
)

plt.ylabel(
    "Residual (Actual - Predicted)"
)

plt.title(
    "PSO Residual Analysis"
)

plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PSO_RESIDUAL_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 72. ACTUAL VS PREDICTED SEQUENCE GRAPH
# =============================================================================

print(
    "Creating pso_actual_predicted_comparison_graph.png..."
)


comparison_df = result_df.sort_values(
    by="Actual_Income"
).reset_index(
    drop=True
)


record_numbers = np.arange(
    len(comparison_df)
)


plt.figure(
    figsize=(
        14,
        7
    )
)


plt.plot(
    record_numbers,
    comparison_df[
        "Actual_Income"
    ],
    linewidth=2,
    label="Actual Income"
)


plt.plot(
    record_numbers,
    comparison_df[
        "PSO_Predicted_Income"
    ],
    linewidth=2,
    label="PSO Predicted Income"
)


plt.xlabel(
    "Validation Records Sorted by Actual Income"
)

plt.ylabel(
    "Farmer Income"
)

plt.title(
    "PSO Actual vs Predicted Farmer Income Comparison"
)

plt.legend()

plt.grid(
    alpha=0.25
)


plt.tight_layout()


plt.savefig(
    PSO_ACTUAL_PRED_GRAPH_PATH,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


# =============================================================================
# 73. TEST PREDICTION SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("PSO TEST PREDICTION SUMMARY")
print("=" * 80)


print(
    f"\nNumber of predictions : "
    f"{len(test_predictions):,}"
)

print(
    f"Minimum prediction    : "
    f"{np.min(test_predictions):,.2f}"
)

print(
    f"Maximum prediction    : "
    f"{np.max(test_predictions):,.2f}"
)

print(
    f"Mean prediction       : "
    f"{np.mean(test_predictions):,.2f}"
)

print(
    f"Median prediction     : "
    f"{np.median(test_predictions):,.2f}"
)

print(
    f"Standard deviation    : "
    f"{np.std(test_predictions):,.2f}"
)


# =============================================================================
# 74. VERIFY SAVED MODEL
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING SAVED PSO MODEL")
print("=" * 80)


try:

    loaded_model = tf.keras.models.load_model(
        PSO_H5_PATH,
        compile=False
    )


    print(
        "\n[OK] PSO H5 model loaded successfully."
    )

    print(
        "Model input shape:",
        loaded_model.input_shape
    )


except Exception as error:

    print(
        "\n[WARNING] Could not reload H5:"
    )

    print(
        error
    )


# =============================================================================
# 75. VERIFY PKL
# =============================================================================

try:

    with open(
        PSO_PKL_PATH,
        "rb"
    ) as file:

        loaded_package = pickle.load(
            file
        )


    print(
        "[OK] PSO preprocessing PKL loaded successfully."
    )


    print(
        "Selected features in PKL:",
        len(
            loaded_package[
                "selected_feature_names"
            ]
        )
    )


except Exception as error:

    print(
        "[WARNING] Could not reload PKL:"
    )

    print(
        error
    )


# =============================================================================
# 76. VERIFY ALL GENERATED FILES
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING PSO OUTPUT FILES")
print("=" * 80)


generated_files = [

    PSO_H5_PATH,
    PSO_PKL_PATH,
    PSO_YAML_PATH,
    PSO_JSON_PATH,

    PSO_SELECTED_FEATURES_PATH,
    PSO_METRICS_CSV_PATH,

    PSO_ACCURACY_GRAPH_PATH,
    PSO_COMPARISON_GRAPH_PATH,
    PSO_HEATMAP_PATH,

    PSO_RESULT_CSV_PATH,
    PSO_RESULT_GRAPH_PATH,

    PSO_PREDICTION_CSV_PATH,
    PSO_PREDICTION_GRAPH_PATH,

    PSO_HISTORY_GRAPH_PATH,
    PSO_CONVERGENCE_GRAPH_PATH,
    PSO_FEATURE_GRAPH_PATH,
    PSO_RESIDUAL_GRAPH_PATH,
    PSO_ACTUAL_PRED_GRAPH_PATH
]


for path in generated_files:

    if os.path.isfile(path):

        file_size_kb = (
            os.path.getsize(
                path
            )
            /
            1024
        )


        print(
            f"[OK] "
            f"{os.path.basename(path)} "
            f"({file_size_kb:,.2f} KB)"
        )

    else:

        print(
            f"[ERROR] "
            f"{os.path.basename(path)}"
        )


# =============================================================================
# 77. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)


print(
    f"\nFeatures before PSO       : "
    f"{TOTAL_FEATURES}"
)

print(
    f"PSO candidate features    : "
    f"{PSO_CANDIDATE_POOL_SIZE}"
)

print(
    f"PSO selected features     : "
    f"{len(selected_feature_names)}"
)

print(
    f"Best PSO fitness          : "
    f"{global_best_score:.6f}"
)


print(
    "\nVALIDATION PERFORMANCE"
)


print(
    f"MAE                      : "
    f"{valid_mae:,.4f}"
)

print(
    f"MSE                      : "
    f"{valid_mse:,.4f}"
)

print(
    f"RMSE                     : "
    f"{valid_rmse:,.4f}"
)

print(
    f"R²                       : "
    f"{valid_r2:.6f}"
)

print(
    f"Explained Variance       : "
    f"{valid_ev:.6f}"
)

print(
    f"Regression Score         : "
    f"{valid_score * 100:.2f}%"
)


print(
    "\nFiles saved in:"
)

print(
    PROJECT_DIR
)


print(
    """
GENERATED PSO FILES
-------------------

1.  pso_model.h5
2.  pso_preprocessor.pkl
3.  pso_config.yaml
4.  pso_metadata.json

5.  pso_selected_features.csv
6.  pso_model_metrics.csv

7.  pso_accuracy_graph.png
8.  pso_comparison_graph.png
9.  pso_heatmap.png

10. pso_result.csv
11. pso_result_graph.png

12. pso_prediction.csv
13. pso_prediction_graph.png

14. pso_training_history_graph.png
15. pso_convergence_graph.png
16. pso_feature_selection_graph.png
17. pso_residual_graph.png
18. pso_actual_predicted_comparison_graph.png
"""
)


print("=" * 80)
print("DONE")
print("=" * 80)
```
